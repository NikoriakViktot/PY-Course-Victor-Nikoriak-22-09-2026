## 1. Сильний пункт CV

З туторіалу (`CV_mini_tutorial_UA.md`, розділ 2):

```text
Сильне дієслово + що зроблено + чим/на яких даних + для чого/яка цінність
```

Слабкий пункт описує **процес** («Responsible for…», «Worked on…»), сильний — **результат**.

### Вправа 1

Напиши `weak_reasons(bullet)` → список причин, чому пункт слабкий (порожній — сильний):

- `"слабкий початок"` — починається з `Responsible for`, `Worked on`, `Helped`, `Participated in`, `Was involved in`, `Did`;
- `"немає дієслова дії"` — перше слово не з `STRONG_VERBS`;
- `"задовгий"` — понад 30 слів;
- `"немає інструментів"` — не згадано жодної технології з `TOOLS`.

````python
STRONG_VERBS = {"Built", "Designed", "Developed", "Automated", "Implemented", "Deployed", "Reduced", "Improved",
                "Migrated", "Containerized", "Tested", "Secured", "Integrated", "Optimized", "Created", "Led"}
WEAK_STARTS = ("responsible for", "worked on", "helped", "participated in", "was involved in", "did")
TOOLS = {"Python", "FastAPI", "Django", "PostgreSQL", "Redis", "Docker", "Compose", "nginx", "GitHub Actions",
         "pytest", "mypy", "SQLAlchemy", "Alembic", "aiogram", "WebSocket", "Channels", "Pydantic", "JWT"}


def weak_reasons(bullet: str) -> list[str]:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    reasons = []
    text = bullet.strip()
    if text.lower().startswith(WEAK_STARTS):
        reasons.append("слабкий початок")
    if text.split()[0] not in STRONG_VERBS:
        reasons.append("немає дієслова дії")
    if len(text.split()) > 30:
        reasons.append("задовгий")
    if not any(tool in text for tool in TOOLS):
        reasons.append("немає інструментів")
    return reasons
    # END SOLUTION


samples = [
    "Responsible for backend development.",
    "Worked on a news project with FastAPI.",
    "Built a news aggregator API with FastAPI, PostgreSQL and Redis, covered by 343 pytest tests.",
    "Containerized two web applications with Docker and Compose behind nginx.",
]
for bullet in samples:
    print(f"{str(weak_reasons(bullet)):<60} {bullet}")
assert weak_reasons(samples[0]) == ["слабкий початок", "немає дієслова дії", "немає інструментів"]
assert weak_reasons(samples[1]) == ["слабкий початок", "немає дієслова дії"]
assert weak_reasons(samples[2]) == [] and weak_reasons(samples[3]) == []
print("✅ Вправа 1 пройдена")
````

## 2. Лише чесні числа

Туторіал (розділ 4): метрики — **тільки справжні**. Число в CV — це обіцянка, що ти зможеш його пояснити на співбесіді.

### Вправа 3

Напиши `unverified_numbers(bullet, facts)` → числа з пункту, яких немає серед значень `facts` (усіх проєктів). Слова-числа й версії на кшталт `3.13` не рахуй — лише цілі числа.

🔮 **Прогноз:** які числа в `CLAIMS` не пройдуть перевірку?

````python
import re

CLAIMS = [
    "Built a news aggregator API with FastAPI, covered by 343 pytest tests.",
    "Improved API performance by 70% with Redis caching.",
    "Automated CI with GitHub Actions: 6 jobs per pull request on 2 Python versions.",
    "Deployed a Django chat app to 2 replicas serving 10000 users.",
]


def unverified_numbers(bullet: str, facts: dict) -> list[int]:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    known = {value for project in facts.values() for value in project.values()}
    numbers = [int(n) for n in re.findall(r"(?<![\d.])\d+(?![\d.])", bullet)]
    return [n for n in numbers if n not in known]
    # END SOLUTION


for claim in CLAIMS:
    print(f"{str(unverified_numbers(claim, FACTS)):<10} {claim}")
assert [unverified_numbers(c, FACTS) for c in CLAIMS] == [[], [70], [], [10000]]
print("✅ Вправа 3 пройдена")
````

<summary>Чому 70% і 10000 — погано</summary>

Кеш Redis у `news_hub` справді є (урок 40), але приріст швидкості ніхто не міряв, а користувачів у навчального проєкту немає. Перше ж питання «як ви це виміряли?» — і довіри до решти CV теж немає. Чесні альтернативи: «with Redis cache-aside and versioned invalidation» (що зроблено) або число, яке ти справді виміряв сам.

## 3. CV як дані → HTML

У проєкті `cv_maker` CV — готовий HTML. Якщо CV кількох версій (Python Developer, Data, академічне), зручніше тримати **дані** окремо й генерувати HTML. Але текст, вставлений у HTML, треба екранувати.

### Вправа 4

Напиши `render(cv)` → HTML-рядок: `<h1>` з ім'ям, для кожного розділу `<h2>` і `<ul>` з пунктами `<li>`. Весь текст із `cv` — через `html.escape`.

````python
import html
from html.parser import HTMLParser

CV = {
    "name": "Olena Koval",
    "sections": {
        "Skills": ["Python, FastAPI, Django", "C++ & Rust <basics>"],
        "Projects": STRONG,
    },
}


def render(cv: dict) -> str:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    parts = [f"<h1>{html.escape(cv['name'])}</h1>"]
    for title, items in cv["sections"].items():
        parts.append(f"<h2>{html.escape(title)}</h2><ul>")
        parts += [f"<li>{html.escape(item)}</li>" for item in items]
        parts.append("</ul>")
    return "\n".join(parts)
    # END SOLUTION


class TextCollector(HTMLParser):
    def __init__(self) -> None:
        super().__init__()
        self.items: list[str] = []
        self.tags: list[str] = []

    def handle_starttag(self, tag, attrs):
        self.tags.append(tag)

    def handle_data(self, data):
        if self.tags and self.tags[-1] == "li":
            self.items.append(data)

    def handle_endtag(self, tag):
        self.tags.pop()


page = render(CV)
parser = TextCollector()
parser.feed(page)
print(page[:300])
assert parser.items[1] == "C++ & Rust <basics>"            # браузер покаже рівно те, що в даних
print("✅ Вправа 4 пройдена")
````

### Вправа 5. Знайди помилку

Та сама функція без `html.escape`. 🔮 **Прогноз:** що побачить рекрутер у пункті `C++ & Rust <basics>`?

Збери `broken_items` — пункти, які отримає парсер (як браузер) зі сторінки без екранування.

````python
def render_unsafe(cv: dict) -> str:
    parts = [f"<h1>{cv['name']}</h1>"]
    for title, items in cv["sections"].items():
        parts.append(f"<h2>{title}</h2><ul>")
        parts += [f"<li>{item}</li>" for item in items]
        parts.append("</ul>")
    return "\n".join(parts)


# YOUR CODE HERE
# BEGIN SOLUTION
unsafe = TextCollector()
unsafe.feed(render_unsafe(CV))
broken_items = unsafe.items
# END SOLUTION

print(broken_items[:3])
assert broken_items[1] == "C++ & Rust " and "<basics>" not in "".join(broken_items)
print("✅ Вправа 5 пройдена")
````

<details>
<summary>Відповідь</summary>

`<basics>` браузер сприйняв як тег і не показав — від пункту лишилось «C++ & Rust ». Той самий клас помилки, що з ім'ям користувача в HTML-повідомленні бота (урок 48): усе, що вставляється в розмітку, — через екранування. Для CV це «лише» зіпсований текст; для сайту, куди текст пише будь-хто, — XSS.
</details>

## 4. PDF

Два генератори з папки проєкту:

| | `generate_cv_pdf.py` (стартовий код) | `generate_cv_pdf_chromium.py` (новий) |
|---|---|---|
| рушій | `wkhtmltopdf` (QtWebKit, проєкт архівований) | Chromium через Playwright |
| встановлення | `pip install pdfkit` + системна програма | `pip install playwright && playwright install chromium` |
| CV викладача / шаблон | 5 / 1 сторінка | 6 / 2 сторінки |

Різниця в сторінках — не помилка: `wkhtmltopdf` за замовчуванням зменшує сторінку, щоб вмістити більше (smart shrinking). З `--disable-smart-shrinking` він дає ті самі 6 і 2 сторінки, що Chromium.

### Вправа 6

Збережи свій CV (`render(CV)`, обгорнутий у `<!DOCTYPE html>…<meta charset="UTF-8">…`) у `my_cv.html`. Далі клітинка спробує зробити PDF тим генератором, що є в середовищі (у Colab — встанови Playwright першою командою з таблиці). Перевір `html_ok`: файл існує і містить `&amp;` (екранований `&`).

````python
# YOUR CODE HERE
# BEGIN SOLUTION
Path("my_cv.html").write_text(
    '<!DOCTYPE html><html><head><meta charset="UTF-8"><title>CV</title></head><body>' + render(CV) + "</body></html>",
    encoding="utf-8")
# END SOLUTION
html_ok = Path("my_cv.html").exists() and "&amp;" in Path("my_cv.html").read_text(encoding="utf-8")
assert html_ok
print("✅ Вправа 6 пройдена")

import importlib.util
import shutil
import sys

if importlib.util.find_spec("playwright"):
    result = subprocess.run([sys.executable, "generate_cv_pdf_chromium.py", str(Path("my_cv.html").resolve()), "my_cv.pdf"],
                            capture_output=True, text=True)
    print(result.stdout or result.stderr[-500:])
elif shutil.which("wkhtmltopdf") and importlib.util.find_spec("pdfkit"):
    print(subprocess.run([sys.executable, "generate_cv_pdf.py", str(Path("my_cv.html").resolve()), "my_cv.pdf"],
                         capture_output=True, text=True).stdout)
else:
    print("PDF-генератора тут немає. Colab: !pip install -q playwright && !playwright install --with-deps chromium")
````
