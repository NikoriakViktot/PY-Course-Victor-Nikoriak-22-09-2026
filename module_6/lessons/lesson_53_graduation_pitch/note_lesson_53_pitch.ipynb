## 1. 5 «чому» як дерево

Тайічі Оно, творець системи виробництва Toyota, радив питати «чому» п'ять разів. Пастка методу — думати, що причина одна. Насправді на кожне «чому» буває кілька відповідей, і це **дерево**: вузол — твердження, діти — відповіді на «чому?». **Корені** — листки дерева: причини у світі, які можна змінити. Корінь — не «немає мого продукту»: це рішення, переодягнене в причину.

````python
WHY = {
    "пропускає важливі новини": ["переглядає новини раз на день", "сповіщення каналів тонуть у чатах", "«неуважна»"],
    "переглядає новини раз на день": ["її теми розкидані по різних рубриках"],
    "її теми розкидані по різних рубриках": ["новини впорядковані за рубриками редакції, а не за темами читача"],
    "сповіщення каналів тонуть у чатах": ["кожен канал сповіщає однаково гучно"],
}
LABELS = {"«неуважна»"}                                   # ярлик, а не причина: змінити нічого
````

### Вправа 1

Напиши `roots(tree, node, labels)` → список листків, досяжних з `node` (у порядку обходу), без ярликів з `labels`. І `path(tree, start, goal)` → список вузлів від `start` до `goal` (ланцюжок «чому»), або `[]`, якщо дороги немає.

````python
def roots(tree: dict[str, list[str]], node: str, labels: set[str]) -> list[str]:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    children = tree.get(node, [])
    if not children:
        return [] if node in labels else [node]
    return [leaf for child in children for leaf in roots(tree, child, labels)]
    # END SOLUTION


def path(tree: dict[str, list[str]], start: str, goal: str) -> list[str]:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    if start == goal:
        return [start]
    for child in tree.get(start, []):
        rest = path(tree, child, goal)
        if rest:
            return [start] + rest
    return []
    # END SOLUTION


found = roots(WHY, "пропускає важливі новини", LABELS)
print("корені:", found)
chain = path(WHY, "пропускає важливі новини", found[0])
print(" → чому? → ".join(chain), f"({len(chain) - 1} «чому»)")
assert found == ["новини впорядковані за рубриками редакції, а не за темами читача", "кожен канал сповіщає однаково гучно"]
assert len(chain) == 4 and path(WHY, "«неуважна»", found[0]) == []
print("✅ Вправа 1 пройдена")
````

## 2. Питання за The Mom Test

Три правила Роба Фіцпатріка: говори про **їхнє життя**, а не свою ідею; питай про **конкретне минуле**, а не майбутнє; менше говори. Погане питання легко впізнати за словами.

### Вправа 2

Напиши `problems(question)` → множину причин, чому питання погане:

- `"майбутнє"` — є частка «б» / «би» окремим словом (умовний спосіб: «користувались би», «ви б платили»);
- `"твоя ідея"` — згадано продукт: слова «бот», «додаток», «наш», «мій»;
- `"навідне"` — питання починається з «Це ж» або містить «правда ж».

Добре питання повертає порожню множину.

````python
import re

QUESTIONS = {
    "Ви б користувались ботом з новинами?": {"майбутнє", "твоя ідея"},
    "Це ж проблема — пропускати новини?": {"навідне"},
    "Скільки ви платили б за наш додаток?": {"майбутнє", "твоя ідея"},
    "Як ви дізналися про останню важливу новину своєї галузі?": set(),
    "Скільки часу вчора пішло на перегляд новин?": set(),
    "За які сервіси для роботи ви вже платите?": set(),
    "Мій продукт вам допоміг би, правда ж?": {"майбутнє", "твоя ідея", "навідне"},
}


def problems(question: str) -> set[str]:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    words = re.findall(r"[\w'’]+", question.lower())
    found = set()
    if {"б", "би"} & set(words):
        found.add("майбутнє")
    if {"бот", "ботом", "бота", "додаток", "наш", "мій", "продукт"} & set(words):
        found.add("твоя ідея")
    if question.lower().startswith("це ж") or "правда ж" in question.lower():
        found.add("навідне")
    return found
    # END SOLUTION


for question, expected in QUESTIONS.items():
    got = problems(question)
    print(("✓" if got == expected else "✗"), f"{question:<58}", sorted(got) or "добре")
    assert got == expected
print("✅ Вправа 2 пройдена")
````

Правила-слова — лише перший фільтр. Справжня перевірка — відповідь: якщо людина розповідає, що вона **робила** («вчора відкрила шість сайтів, а новину про тариф побачила в обід»), питання добре. Якщо каже, що **зробила б** («так, цікаво, користувалась би»), — це компліменти, а не дані.

## 3. Ринок знизу вгору

TAM — усі, хто має проблему; SAM — кого може обслужити продукт; SOM — кого досягнеш за рік. Дохід = SOM × ціна × 12. Кожне число — або з джерелом, або **чесно позначене припущення**.

### Вправа 3

Напиши `bottom_up(inputs)`. Кожен вхід — пара `(значення, джерело)`, де джерело `None` означає «припущення». Поверни словник з `TAM`, `SAM`, `SOM` (цілі, округлені вниз), `дохід_рік` і `припущення` — відсортований список назв входів без джерела.

````python
import math

INPUTS = {
    "фахівців PR, маркетингу, аналітики": (40_000, None),
    "частка, що читає новини в Telegram": (0.72, "Internews, 2025"),
    "частка SAM, яку досягнемо за рік": (0.01, None),
    "ціна за місяць, грн": (99, None),
}


def bottom_up(inputs: dict[str, tuple[float, str | None]]) -> dict[str, object]:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    people, telegram, reach, price = (value for value, _ in inputs.values())
    tam = math.floor(people)
    sam = math.floor(tam * telegram)
    som = math.floor(sam * reach)
    return {"TAM": tam, "SAM": sam, "SOM": som, "дохід_рік": som * price * 12,
            "припущення": sorted(name for name, (_, source) in inputs.items() if source is None)}
    # END SOLUTION


market = bottom_up(INPUTS)
for key, value in market.items():
    print(f"{key:<12} {value}")
assert (market["TAM"], market["SAM"], market["SOM"]) == (40_000, 28_800, 288)
assert market["дохід_рік"] == 288 * 99 * 12 == 342_144
assert len(market["припущення"]) == 3
print("✅ Вправа 3 пройдена")
````

🔮 **Прогноз:** три з чотирьох чисел — припущення. Чи варто тоді взагалі ставити цей слайд?

Так — але з позначками. Інвестор не чекає точності від проєкту без користувачів; він перевіряє, чи ти **вмієш рахувати** і чи **відрізняєш** знання від здогадок. 40 000 фахівців — число, яке ти перевіриш першим; 1% — найбільш сумнівне, і саме його підтвердить або спростує пілот. Слабкий слайд — «ринок — мільярди доларів, візьмемо 1%»: там немає жодної перевірюваної частини.

### Вправа 4

Напиши `seconds_per_slide(slides)` → список `(назва, секунд)`: скільки триває мова на кожному слайді за нотатками при темпі `WORDS_PER_MINUTE` (округли до цілих). Потім знайди `longest` — назву слайда з найдовшою мовою.

🔮 **Прогноз:** який слайд прикладу найдовший?

````python
def seconds_per_slide(slides) -> list[tuple[str, int]]:
    # YOUR CODE HERE
    # BEGIN SOLUTION
    return [(s.title, round(words(s.notes) / WORDS_PER_MINUTE * 60)) for s in slides]
    # END SOLUTION


timing = seconds_per_slide(slides)
# YOUR CODE HERE
# BEGIN SOLUTION
longest = max(timing, key=lambda item: item[1])[0]
# END SOLUTION

for title, seconds in timing:
    print(f"{title:<28} {seconds:>4} с  " + "▇" * (seconds // 5))
total = sum(seconds for _, seconds in timing)
print(f"разом мова: {total // 60} хв {total % 60} с; найдовший: {longest}")
assert longest == "Проблема" and 4 * 60 <= total <= 7 * 60
print("✅ Вправа 4 пройдена")
````

Проблема — найдовший слайд, і це правильно: на ньому вирішується, чи слухатимуть далі. Демо коротке за словами, але довге за часом — там ти **показуєш**, а не говориш. Тому мова в прикладі — ~4 хвилини, а разом з демо — 5–7.

### Вправа 5

Запусти `check()` на `DRAFT` і збери в `fixes` множину **номерів слайдів**, які треба переробити (з повідомлень виду `слайд N «…»`).

````python
DRAFT = """---
marp: true
---
# Мій фінальний проєкт

<!-- Привіт, я розкажу про свій проєкт. -->
---
# Технології

Python 3.13, FastAPI, SQLAlchemy 2, Alembic, PostgreSQL 16, Redis 7, aiogram 3, Docker, Docker Compose,
nginx, GitHub Actions, pytest, mypy, httpx, Pydantic 2, Gemini API, Anthropic API, bcrypt, PyJWT, Marp,
BeautifulSoup, aiohttp, uvicorn, fakeredis, factory, Playwright, Channels, daphne, Django 5.2, DRF,
SQLite, Jupyter, Colab, pandas, Dash, Streamlit, gRPC, Postman, Linux, bash.

<!-- Ось усі технології, які я використав. -->
---
# Ринок

Ринок — мільярди доларів, візьмемо [X]%.

<!-- TODO -->
---
# Дякую за увагу!

<!-- Питання? -->
"""

problems_found = check(DRAFT)
for problem in problems_found:
    print("✗", problem)

# YOUR CODE HERE
# BEGIN SOLUTION
fixes = {int(m.group(1)) for p in problems_found if (m := re.match(r"слайд (\d+)", p))}
# END SOLUTION

print("переробити слайди:", sorted(fixes))
assert fixes == {2, 3}
assert "немає слайда «проблема»" in problems_found and "слайдів 4, треба 8–12" in problems_found
print("✅ Вправа 5 пройдена")
````
