## 1. Як Python виконує код

Коли ти пишеш файл `script.py` і запускаєш його, відбувається не "магія", а конкретний процес:

```
джерельний код (.py) → байткод → віртуальна машина Python
```

Python читає твій файл, перекладає його у проміжну форму (**байткод**) і виконує цю форму на віртуальній машині. Це причина, чому поруч зі скриптом іноді з'являється папка `__pycache__` — там Python кешує байткод.

### "Interpreted" чи "compiled"?

Python зазвичай називають **інтерпретованою мовою** — не тому, що компіляції немає (вона є, у байткод), а тому що ти запускаєш програму просто з джерельного файлу, без окремого кроку "спочатку скомпілювати в exe". Правильніше думати про це не як про бінарний вибір, а як про спектр: швидкість зворотного зв'язку проти швидкості виконання.

## 2. Два режими роботи: REPL і скрипт

### REPL (Read-Eval-Print Loop)

Коли ти запускаєш `python3` (або `py` на Windows) без файлу — відкривається інтерактивний режим із запрошенням `>>>`. Кожен рядок виконується **одразу**, і Python сам показує результат.

```
>>> 2 + 2
4
>>> "hello"
'hello'
```

Вийти: `exit()`, або `Ctrl+D` (macOS/Linux) / `Ctrl+Z` потім `Enter` (Windows).

### Скрипт (.py файл)

Файл виконується **цілком, згори донизу**, і **нічого не друкує сама по собі** — доки ти явно не викличеш `print()`.

## 3. `print()`, `sep`, `end`, `\n`, `\t`

`print()` приймає кілька аргументів і два іменовані параметри, які керують тим, як саме вони з'являються на екрані:

- `sep` — що ставити **між** аргументами (за замовчуванням — пробіл);
- `end` — що ставити **в кінці** виклику (за замовчуванням — перехід на новий рядок `\n`).

Escape-послідовності всередині рядка:

- `\n` — новий рядок;
- `\t` — символ табуляції (вирівнює до наступної позиції, кратної 8, а не просто «додає 4 пробіли» — це важливо в домашньому завданні).


````python
print("one", "two", "three")                 # sep за замовчуванням — пробіл
print("one", "two", sep=", ")                 # sep замінює пробіл між аргументами
print("one", "two", end="...")                # end замінює перехід на новий рядок
print()                                        # порожній print — просто переносить рядок
print("A\tB")                                  # таб між A і B
print("A\nB")                                  # новий рядок між A і B
````
