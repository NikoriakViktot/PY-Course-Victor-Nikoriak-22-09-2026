## 1. Список — замовлення столика живе

**Проблема.** Столик 4 замовив борщ, вареники й каву. Потім гість дозамовив узвар, передумав щодо кави, а кухня подає хліб першим. Замовлення **змінюється весь вечір**, і порядок у ньому важливий: що подавати першим.

**Дані.** Кілька назв страв у певному порядку, кількість наперед невідома.

**Тип.** `list` — впорядкований набір, який можна змінювати на місці.


### Створити список

Список записують у **квадратних дужках**, елементи — через кому. Список:

- **зберігає порядок** — страви стоять так, як ми їх записали;
- **дозволяє повтори** — дві кави будуть двома окремими елементами;
- може містити будь-які значення, навіть різних типів.


````python
items = ["борщ", "вареники", "кава"]
print(items)
print(type(items))
````

### `len()` — скільки елементів

Вбудована функція `len()` повертає кількість елементів. Вона знадобиться в кожному циклі: останній індекс списку завжди `len(items) - 1`.


````python
print("Позицій у замовленні:", len(items))
print("Останній індекс:", len(items) - 1)
````

### Індекс — елемент за номером позиції

До елемента звертаються за **індексом** у квадратних дужках. Нумерація починається з **0**, як у рядках з уроку 3. Від'ємний індекс рахує з кінця: `-1` — останній.

```text
items:      ["борщ", "вареники", "кава"]
індекс +:       0         1         2
індекс −:      -3        -2        -1
```


````python
print("Перша страва:  ", items[0])
print("Друга страва:  ", items[1])
print("Остання страва:", items[-1])
````

### 🔮 Індекс за межами списку

У списку три елементи, тобто індекси 0, 1, 2. Що станеться з `items[3]`?

Python не «вигадує» елемент — він зупиняє програму з `IndexError`. Тут ми ловимо помилку через `try/except`, щоб побачити її текст і продовжити.


````python
try:
    print(items[3])
except IndexError as e:
    print("IndexError:", e)
````

### `in` — чи є значення в списку

Оператор `in` повертає `True` або `False`. Це найпростіший спосіб перевірити наявність перед дією, яка може впасти, наприклад перед `.remove()`.


````python
print("Є кава?", "кава" in items)
print("Є піца?", "піца" in items)
print("Піци немає?", "піца" not in items)
````

### Замінити елемент за індексом

Список **змінний**: можна записати нове значення на місце старого — `items[i] = нове`. Новий список при цьому не створюється, змінюється той самий.

Гість попросив замість вареників деруни:


````python
items[1] = "деруни"
print(items)
items[1] = "вареники"     # повертаємо як було — щоб далі історія йшла від однакового стану
print(items)
````

### Вечір іде — замовлення змінюється

Кожна подія вечора — один метод списку. Метод викликають через крапку: `items.append(...)`. Запам'ятай пару «подія → метод»:

| Подія в кафе | Метод |
|---|---|
| гість дозамовив узвар (у кінець) | `.append("узвар")` |
| кухня подає хліб першим (на початок) | `.insert(0, "хліб")` |
| гість відмовився від кави | `.remove("кава")` |
| страву винесли — забираємо з черги | `.pop()` / `.pop(0)` |
| сусідній столик приєднався зі своїм замовленням | `.extend([...])` |
| скільки разів кава / де вона стоїть | `.count("кава")`, `.index("кава")` |
| стіл розрахувався — замовлення порожнє | `.clear()` |


#### Старт: замовлення столика 4

Гість замовив три страви. Створюємо список заново, щоб усі кроки нижче йшли від однакового стану. Якщо запускаєш клітинки повторно, починай із цієї.

```text
індекс:    0        1          2
        ["борщ", "вареники", "кава"]
```


````python
items = ["борщ", "вареники", "кава"]
print("старт  :", items)
````

#### `.append(x)` — додати в кінець

Гість дозамовив узвар. Нова страва стає **останньою** в черзі: `.append()` додає **один** елемент у кінець списку.

- Змінює сам список на місці, нового списку не створює.
- Повертає `None`. Тому `items = items.append("узвар")` — помилка: список втратиться.

```text
["борщ", "вареники", "кава"]  →  ["борщ", "вареники", "кава", "узвар"]
                                                              ↑ новий
```


````python
items.append("узвар")            # дозамовили — у кінець
print("append :", items)
````

#### `.insert(i, x)` — вставити на позицію

Кухня подає хліб першим, тому він має стати на **початок** списку. `.insert(i, x)` ставить `x` на індекс `i`, а всі елементи від цієї позиції зсуваються на одну праворуч.

- `.insert(0, x)` — на початок; `.insert(len(items), x)` — те саме, що `.append(x)`.
- Індекс за межами не дає помилки: `.insert(100, x)` просто додасть у кінець.

```text
          ↓ індекс 0
["хліб", "борщ", "вареники", "кава", "узвар"]
          └──── усі зсунулись на +1 ────┘
```


````python
items.insert(0, "хліб")          # хліб першим — на позицію 0
print("insert :", items)
````

#### `.remove(x)` — видалити за значенням

Гість відмовився від кави. Ми знаємо **що** прибрати, але не знаємо **де** воно стоїть. `.remove(x)` шукає значення сам і видаляє **перше** входження.

- Якщо кав дві, зникне лише перша.
- Якщо значення немає, буде `ValueError` (наступна клітинка).
- Видалити за **позицією**, а не за значенням, можна через `.pop(i)` або `del items[i]`.


````python
items.remove("кава")             # відмовились — видаляє ПЕРШЕ входження значення
print("remove :", items)
````

#### 🔮 `.remove()` значення, якого немає

Що станеться, якщо викликати `items.remove("піца")`, коли піци в замовленні немає?

Програма зупиниться з `ValueError`. Правильний шаблон — спершу перевірити `in`: тоді видалення відбувається лише тоді, коли є що видаляти.


````python
try:
    items.remove("піца")
except ValueError as e:
    print("ValueError:", e)

if "піца" in items:              # безпечний шаблон
    items.remove("піца")
print("items без змін:", items)
````

#### `.pop(i)` — забрати за позицією і повернути

Хліб винесли до столу, тож його треба прибрати із черги. При цьому ми хочемо знати, **що саме** винесли. `.pop()` — єдиний із цих методів, який **повертає** видалений елемент.

- `.pop()` без аргументу забирає **останній** елемент, а `.pop(0)` — перший.
- Порожній список: `.pop()` дає `IndexError: pop from empty list`.

```text
served = items.pop(0)
["хліб", "борщ", "вареники", "узвар"]  →  ["борщ", "вареники", "узвар"],   served = "хліб"
```


````python
served = items.pop(0)            # хліб винесли — забираємо з початку і повертаємо
print("pop(0) :", items, "   винесли:", served)
````

#### `.extend([...])` — додати кілька елементів

Сусідній столик приєднався і додав своє: два чаї. `.extend()` бере **кожен** елемент переданої послідовності і додає його в кінець.

Порівняй з `.append()`: він додав би весь список **одним** елементом.

```python
items.extend(["чай", "чай"])   # [..., "чай", "чай"]     — два нові елементи
items.append(["чай", "чай"])   # [..., ["чай", "чай"]]   — один елемент-список усередині
```


````python
items.extend(["чай", "чай"])     # сусідній столик додав своє
print("extend :", items)
````

#### `.count(x)` — скільки разів трапляється

Кухня питає, скільки чаїв готувати. `.count(x)` нічого не змінює в списку — лише рахує входження. Якщо значення немає, повертає `0`, а не помилку.


````python
print("Чаю замовлено:", items.count("чай"))
print("Кави замовлено:", items.count("кава"))
````

#### `.index(x)` — де стоїть перше входження

Який чай подати першим? `.index(x)` повертає індекс **першого** входження. На відміну від `.count()`, для відсутнього значення він дає `ValueError`.


````python
print("Перший чай на індексі:", items.index("чай"))
try:
    items.index("кава")
except ValueError as e:
    print("ValueError:", e)
````

#### `del items[i]` — видалити за індексом без повернення

`del` — не метод, а інструкція Python. Вона видаляє елемент на позиції, але, на відміну від `.pop(i)`, нічого не повертає. Працює і зі зрізом: `del items[1:3]`.


````python
del items[-1]                    # один чай прибрали — другий лишився
print("del    :", items)
````

#### `.clear()` — спорожнити список

Стіл розрахувався, замовлення закрите. `.clear()` видаляє всі елементи, але сам список лишається — порожнім, з тим самим ім'ям.

| Метод | Змінює список? | Повертає |
|---|---|---|
| `.append(x)`, `.insert(i, x)`, `.extend(...)`, `.remove(x)`, `.clear()` | так | `None` |
| `.pop(i)` | так | видалений елемент |
| `.count(x)`, `.index(x)` | ні | число |


````python
items.clear()
print("clear  :", items, "  довжина:", len(items))
````

### Зріз — частина списку

**Зріз** `items[start:stop]` повертає **новий** список з елементів від `start` до `stop`, **не включаючи** `stop`. Сам список не змінюється.

```text
items:   ["хліб", "борщ", "вареники", "узвар", "чай"]
індекс:      0       1        2          3       4
items[1:3]  →        ["борщ", "вареники"]
```


````python
items = ["хліб", "борщ", "вареники", "узвар", "чай"]
print("items[1:3] =", items[1:3])
print("items      =", items, "  ← без змін")
````

### Зріз без краю

Пропущений `start` означає «з початку», пропущений `stop` — «до кінця». Так зручно брати перші або останні кілька елементів.


````python
print("перші дві:     items[:2] =", items[:2])
print("від 2 до кінця: items[2:] =", items[2:])
print("останні дві:  items[-2:] =", items[-2:])
````

### Зріз з кроком

Третє число — **крок**: `items[::2]` бере кожен другий елемент. Від'ємний крок іде у зворотному порядку: `items[::-1]` — новий список задом наперед.


````python
print("кожна друга: items[::2]  =", items[::2])
print("навпаки:     items[::-1] =", items[::-1])
````

### Зріз не боїться меж

На відміну від індексу, зріз за межами списку **не дає помилки** — просто повертає те, що є. А `items[:]` — повна копія списку.


````python
print("items[1:100] =", items[1:100])
print("items[10:]   =", items[10:])
print("копія items[:] =", items[:])
````

### `sorted()` — новий впорядкований список

Власниця хоче бачити ціни страв від дешевої до дорогої. Вбудована функція `sorted()` повертає **новий** список, а початковий не чіпає. `reverse=True` — від більшого до меншого.


````python
prices = [55, 70, 40, 55]
print("sorted(prices)               =", sorted(prices))
print("sorted(prices, reverse=True) =", sorted(prices, reverse=True))
print("prices                       =", prices, "  ← оригінал без змін")
````

### `.sort()` — впорядкувати сам список

Метод `.sort()` змінює **сам** список на місці і повертає `None`. Обирай його, коли старий порядок більше не потрібен.


````python
prices.sort()
print("після .sort():", prices)
````

### 🔮 Пастка: `prices = prices.sort()`

Що буде в `prices` після такого рядка?

`.sort()` повертає `None`, і саме `None` записується в змінну — список втрачено. Правильно: або `prices.sort()` окремим рядком, або `prices = sorted(prices)`.


````python
oops = [3, 1, 2]
oops = oops.sort()
print("oops =", oops)
````

### `.reverse()` — розвернути на місці

`.reverse()` розвертає порядок самого списку (повертає `None`). `items[::-1]` робить те саме, але як **новий** список.


````python
queue = ["хліб", "борщ", "узвар"]
queue.reverse()
print("після .reverse():", queue)
````

### `sum()`, `min()`, `max()` — підсумки числового списку

Три вбудовані функції для списку чисел: сума, найменше, найбільше. Для порожнього списку `min()` і `max()` дають `ValueError`, а `sum([])` — `0`.


````python
prices = [40, 55, 55, 70]
print("сума:", sum(prices), " найдешевше:", min(prices), " найдорожче:", max(prices))
print("середня ціна:", sum(prices) / len(prices))
````

### 🔮 Два імені — один список

Адміністраторка робить «копію» замовлення для кухні: `kitchen = items`. Потім кухня дописує туди «хліб». Що надрукує `print(items)`?

`kitchen = items` **не копіює** список — воно прив'язує друге ім'я до того самого об'єкта. Тому зміну видно через обидва імена. `is` перевіряє, чи це один і той самий об'єкт.


````python
items = ["борщ", "вареники"]
kitchen = items                  # НЕ копія: друге ім'я того самого списку
kitchen.append("хліб")
print("kitchen =", kitchen)
print("items   =", items, "  ← змінився теж!")
print("той самий об'єкт?", kitchen is items)
````

### `.copy()` — справжня копія

`.copy()` створює **новий** список з тими самими елементами. Тепер зміни в копії не зачіпають оригінал. Те саме дають `items[:]` і `list(items)`.

```text
  items   ──┐
             ├──►  ['борщ', 'вареники', 'хліб']            один об'єкт, два імені
  kitchen ──┘
  backup  ─────►  ['борщ', 'вареники', 'хліб', 'узвар']    .copy() → окремий об'єкт
```


````python
backup = items.copy()
backup.append("узвар")
print("backup =", backup)
print("items  =", items, "  ← копія живе окремо")
print("той самий об'єкт?", backup is items)
````

### Пройти список циклом `while`

Скільки кав замовив столик? Поки що в нас є лише `while` (урок 4). Заводимо індекс `i`, починаємо з 0 і йдемо, доки `i` менший за довжину списку.

**Три питання перед будь-яким алгоритмом:**
1. Як я отримую кожен елемент? → `items[i]`
2. Що я роблю з ним? → порівнюю з `"кава"`
3. Де зберігаю результат? → у лічильнику `coffee`

```text
            i = 0
              │
   ┌──────────▼──────────┐
   │   i < len(items) ?  │── ні ──► вихід: coffee готовий
   └──────────┬──────────┘
             так
              │
     items[i] == "кава" ? ── так ──► coffee += 1
              │
           i += 1  ──────────────► назад до перевірки
```


````python
items = ["кава", "чай", "кава", "узвар", "кава"]

coffee = 0          # лічильник
i = 0               # індекс
while i < len(items):
    print(f"  i={i}  items[i]={items[i]!r:<9} coffee={coffee}")
    if items[i] == "кава":
        coffee += 1
    i += 1          # ОБОВ'ЯЗКОВО, інакше нескінченний цикл

print("Кав у замовленні:", coffee, "  (готовим методом:", items.count("кава"), ")")
````

🔮 **Передбач:** що станеться з `while i <= len(items):`?

<details><summary>Відповідь</summary>
При <code>i = 5</code> умова <code>5 <= 5</code> ще істинна, а <code>items[5]</code> не існує → <code>IndexError: list index out of range</code>. Остання позиція завжди <code>len(items) - 1</code>.
</details>

Для «скільки разів» є готовий `.count()`. Але **той самий цикл** потрібен для сум, середніх і максимумів, для яких методу немає — побачимо в блоці 4.

### Схема: що вміє список

```text
                         LIST  — впорядкований, змінний, дублікати дозволені
  ┌───────────┬────────────────┬──────────────────┬──────────────────┬──────────────────┐
  │ створити  │ прочитати      │ змінити          │ видалити         │ впорядкувати     │
  ├───────────┼────────────────┼──────────────────┼──────────────────┼──────────────────┤
  │ []        │ items[0]       │ items[i] = x     │ .remove(x)       │ .sort()  (сам)   │
  │ list(...) │ items[-1]      │ .append(x)       │ .pop() / .pop(i) │ sorted() (новий) │
  │ .copy()   │ items[1:3]     │ .insert(i, x)    │ .clear()         │ .reverse()       │
  │           │ len, in        │ .extend([...])   │ del items[i]     │ items[::-1]      │
  │           │ .count .index  │                  │                  │ sum min max      │
  └───────────┴────────────────┴──────────────────┴──────────────────┴──────────────────┘
```


## 2. Кортеж — закритий чек є фактом

**Проблема.** Столик 4 розрахувався: номер столу, сума замовлення, чайові. Чек **закрито** — його вже ніхто не має права «підправити». Якби чек був списком, випадковий `.append()` або `receipt[1] = 600` пройшли б мовчки.

**Дані.** Три значення різного сенсу у фіксованому порядку: позиція 0 — стіл, 1 — сума, 2 — чайові.

**Тип.** `tuple` — впорядкований **незмінний** запис.


### Створити кортеж

Кортеж записують у **круглих дужках**, значення через кому. Кожна позиція має свій сенс, тому порядок полів треба пам'ятати: `(стіл, сума, чайові)`.


````python
receipt = (4, 860.0, 90.0)       # (стіл, сума замовлення, чайові)
print(receipt)
print(type(receipt))
````

### Індекси, зрізи, `len`, `in` — як у списку

Усе, що **читає** дані, працює так само, як у списку. Зріз кортежу — теж кортеж.


````python
print("Стіл:   ", receipt[0])
print("Чайові: ", receipt[-1])
print("Зріз:   ", receipt[1:])
print("Довжина:", len(receipt), "  860.0 є?", 860.0 in receipt)
````

### 🔮 Спроба змінити кортеж

Касир хоче «підправити» суму закритого чека: `receipt[1] = 600.0`. Що станеться?

Кортеж **незмінний** — Python зупиняє спробу з `TypeError`. Саме для цього чек зберігають кортежем: захист від випадкової зміни.


````python
try:
    receipt[1] = 600.0
except TypeError as e:
    print("TypeError:", e)
````

### Методи кортежа: лише `.count()` і `.index()`

У кортежа немає `.append()`, `.remove()`, `.sort()` — жодного методу, що змінює вміст. Лишаються два методи, які тільки читають. Щоб «змінити» кортеж, створюють новий.


````python
print("скільки разів 90.0:", receipt.count(90.0))
print("де стоїть 860.0:   ", receipt.index(860.0))
print("є .append?", hasattr(receipt, "append"))
````

### Пакування — кома створює кортеж

Значення через кому самі складаються в кортеж, навіть **без дужок**. Це називається **пакуванням**. Дужки лише роблять запис наочнішим.


````python
point = 4, 860.0, 90.0
print(type(point).__name__, point)
````

### Розпакування — кортеж у кілька змінних

Каса рахує, скільки гість заплатив разом із чайовими. Замість `receipt[1] + receipt[2]` кортеж **розпаковують**: кожна змінна зліва отримує своє значення справа, за порядком. Код читається як текст, без «магічних» індексів.


````python
table, total, tip = receipt
print(f"Стіл {table}: замовлення {total} + чайові {tip} = до сплати {total + tip} грн")
````

### Обмін значень без тимчасової змінної

Пакування і розпакування разом дають обмін в один рядок: праворуч пакується кортеж `(b, a)`, ліворуч розпаковується в `a, b`.


````python
a, b = 1, 2
a, b = b, a
print("після обміну: a =", a, " b =", b)
````

### 🔮 Розпакування вимагає точної кількості

Що буде, якщо зліва дві змінні, а в кортежі три значення?

Кількість змінних має **дорівнювати** кількості значень, інакше — `ValueError`.


````python
try:
    table, total = receipt
except ValueError as e:
    print("ValueError:", e)
````

### 🔮 `(42)` чи `(42,)`?

Що таке `(42)`? Дужки нічого не роблять — це просто число в дужках, як у математиці. Кортеж створює **кома**, тому кортеж з одного елемента пишуть `(42,)`.


````python
print(type((42)).__name__, "  ← просто число")
print(type((42,)).__name__, "← кортеж з одного елемента")
````

### Незмінний — але що саме?

Кортеж не дає **замінити** свій елемент. Але якщо елемент сам змінний (наприклад, список), його вміст змінити можна. Для повністю незмінних записів клади в кортеж числа, рядки, інші кортежі.


````python
t = (4, ["борщ", "кава"])
t[1].append("чай")
print(t, "  ← кортеж той самий, змінився список усередині")
````

### Схема: життя кортежа

```text
   пакування                 читання                  розпакування
   4, 860.0, 90.0     ──►   receipt[0]  receipt[-1]  ──►  table, total, tip = receipt
                            receipt[1:]  len()  in        a, b = b, a
                            .count() .index()
                            ✗ receipt[1] = ...  → TypeError
```

| | `list` | `tuple` |
|---|---|---|
| Розмір змінюється? | так | ні, форма фіксована |
| Що означає позиція? | усі елементи рівноправні | кожна позиція має свій сенс |
| У кафе | позиції замовлення, чеки дня | один закритий чек |
| Може бути елементом `set`? | ні | так (хешований) |


## 3. NamedTuple — чек з іменами полів

**Проблема.** Через тиждень відкриваєш код: `receipt[1]` — це сума чи чайові? А коли полів стане шість (офіціант, сума, чайові, день, час, кількість гостей) — переплутати індекс легко, і Python **не помітить**: мовчки порахує не те.

**Дані.** Той самий закритий чек, але полів більше, і кожне має зрозумілу назву.

**Тип.** `NamedTuple` — звичайний кортеж, у якого поля мають імена.


### Описати шаблон запису

Спочатку описуємо **шаблон чека** — назву і поля:

- `from typing import NamedTuple` підключає інструмент з модуля `typing` (як `import random` в уроці 4);
- `class Order(NamedTuple):` — назва шаблону;
- кожен рядок усередині — поле і тип, який ми туди **очікуємо**.

Слово `class` належить до теми класів (модуль 2). Поки що читай цей блок як «опис полів запису». Клітинка нічого не друкує — вона лише описує шаблон.


````python
from typing import NamedTuple


class Order(NamedTuple):
    waiter: str          # хто обслуговував
    total_bill: float    # сума чека, грн
    tip: float           # чайові, грн
    day: str             # день тижня
    time: str            # "обід" / "вечеря"
    size: int            # гостей за столом
````

### Створити чек за шаблоном

`Order(...)` створює один чек: значення розкладаються по полях **у порядку оголошення**. На відміну від звичайного кортежу, `print` показує імена полів.


````python
order = Order("Тарас", 540.0, 50.0, "пт", "вечеря", 2)
print(order)
````

### Поле за іменем — замість індексу

До поля звертаються через крапку: `order.tip`. Індекс теж працює, бо це досі кортеж, але ім'я одразу пояснює, що саме ми беремо.

```text
   Order( waiter , total_bill , tip , day , time , size )
            [0]       [1]       [2]   [3]   [4]    [5]
          .waiter  .total_bill  .tip  .day  .time  .size
```


````python
print("order.waiter =", order.waiter)
print("order.tip    =", order.tip)
print("order[2]     =", order[2], "  ← те саме значення")
````

### `NamedTuple` — це справжній кортеж

`isinstance(x, tuple)` перевіряє, чи `x` є кортежем. Для `Order` — так. Тому все, що вміє кортеж (індекси, зрізи, розпакування, незмінність), уміє і `Order`.


````python
print("Order — кортеж?", isinstance(order, tuple))
print("зріз:", order[1:3])
````

### Розпакування чека

Розпаковується так само, як звичайний кортеж — у шість змінних, за порядком полів.


````python
waiter, total_bill, tip, day, time, size = order
print(f"{waiter} обслужив {size} гостей у {day} ({time}): {total_bill} грн + {tip} чайових")
````

### 🔮 Спроба змінити поле

Що станеться з `order.tip = 70.0`?

`NamedTuple` незмінний, як і будь-який кортеж — `AttributeError`.


````python
try:
    order.tip = 70.0
except AttributeError as e:
    print("AttributeError:", e)
````

### `._fields` — які поля в шаблоні

Службові методи `NamedTuple` починаються з підкреслення, щоб не конфліктувати з іменами полів. `._fields` повертає кортеж назв полів — зручно, щоб нагадати собі порядок.


````python
print(Order._fields)
````

### `._asdict()` — пари «ім'я → значення»

`._asdict()` показує чек як пари «назва поля → значення». Це словник — тема уроку 6; тут він лише для зручного перегляду.


````python
print(order._asdict())
````

### `._replace()` — «виправлений» чек як новий

Змінити чек не можна, але можна створити **новий** з іншим значенням поля. `._replace(tip=70.0)` повертає новий `Order`, старий лишається без змін.


````python
fixed = order._replace(tip=70.0)
print("старий:", order.tip, "  новий:", fixed.tip)
print("той самий об'єкт?", fixed is order)
````

🔮 **Передбач:** `Order("Тарас", "п'ятсот", 50.0, "пт", "вечеря", 2)` — буде помилка?

<details><summary>Відповідь</summary>
Ні. Python не перевіряє типи з шаблону: чек створиться, а проблема вилізе пізніше, коли <code>total_bill</code> спробують додати до числа. Типи в шаблоні — підказка для читача коду.
</details>


## 4. Чеки дня — список записів

**Проблема.** Зміна іде: чеки закриваються один за одним, і наприкінці власниця хоче підсумок — скільки чеків, виторг, частка чайових, найбільший чек, скільки було вечерь, які столи великі.

**Дані.** Багато однакових за формою записів `Order`, що з'являються протягом дня.

**Тип.** Список змінний (чеки додаються), а кожен елемент у ньому незмінний (`Order`). Два рівні: `list` з кортежів.


### Збираємо чеки протягом зміни

Зміна починається з **порожнього** списку `[]`. Кожен закритий чек додаємо в кінець через `.append()` — саме тому для чеків дня потрібен список, а не кортеж.


````python
orders = []

orders.append(Order("Тарас",   540.0,  50.0, "пт", "вечеря", 2))
orders.append(Order("Марія",   320.0,  30.0, "пт", "обід",   1))
orders.append(Order("Тарас",   980.0, 120.0, "сб", "вечеря", 4))
orders.append(Order("Олексій", 760.0,  70.0, "сб", "вечеря", 3))
orders.append(Order("Марія",   450.0,   0.0, "нд", "обід",   5))

print("Чеків за зміну:", len(orders))
````

### Два рівні доступу: чек → поле

`orders[2]` — третій чек у списку (індекс списку), а `.total_bill` — поле цього чека (ім'я поля кортежу).

```text
  orders  (list — росте протягом дня)
  ├── [0]  Order("Тарас",   540.0,  50.0, "пт", "вечеря", 2)
  ├── [1]  Order("Марія",   320.0,  30.0, "пт", "обід",   1)
  ├── [2]  Order("Тарас",   980.0, 120.0, "сб", "вечеря", 4)  ← orders[2].total_bill → 980.0
  ├── [3]  Order("Олексій", 760.0,  70.0, "сб", "вечеря", 3)
  └── [4]  Order("Марія",   450.0,   0.0, "нд", "обід",   5)
```


````python
print("Третій чек:  ", orders[2])
print("Його сума:   ", orders[2].total_bill, "  обслужив:", orders[2].waiter)
print("Останній чек:", orders[-1].day, orders[-1].time)
````

### Алгоритм 1 — накопичувач: виторг і чайові

**Накопичувач** — змінна, яка перед циклом дорівнює нулю, а на кожному кроці до неї додається поле поточного чека.

Думаємо перед тим, як кодити: елемент — `orders[i]`; робимо — додаємо його суму й чайові; зберігаємо — у `revenue` і `tips`. Рядок `order = orders[i]` дає коротке ім'я поточному чеку.


````python
revenue = 0
tips = 0

i = 0
while i < len(orders):
    order = orders[i]
    revenue += order.total_bill
    tips += order.tip
    i += 1

print("Виторг:", revenue, "грн")
print("Чайові:", tips, "грн  =", round(tips / revenue * 100, 1), "% від виторгу")
````

### Алгоритм 2 — «поточний чемпіон»: найбільший чек

Пам'ятаємо **найбільший чек, який уже бачили**. Стартуємо з першого чека, а не з нуля: якби всі чеки були поверненнями зі знаком мінус, нуль став би «максимумом», якого немає в даних. Тому цикл починається з `i = 1`. Чемпіона оновлюємо лише тоді, коли знайшли більший.

Зберігаємо весь чек, а не лише суму, — щоб потім назвати офіціанта.

```text
  best = orders[0] (540.0)
     i=1  320.0 > 540.0 ?  ні   → best 540.0
     i=2  980.0 > 540.0 ?  ТАК  → best 980.0  (Тарас)
     i=3  760.0 > 980.0 ?  ні   → best 980.0
     i=4  450.0 > 980.0 ?  ні   → best 980.0
```


````python
best = orders[0]
i = 1
while i < len(orders):
    if orders[i].total_bill > best.total_bill:
        print(f"  i={i}: {orders[i].total_bill} > {best.total_bill} → новий чемпіон {orders[i].waiter}")
        best = orders[i]
    i += 1

print(f"Найбільший чек: {best.total_bill} грн, обслужив {best.waiter}")
````

### Алгоритм 3 — лічильник: скільки було вечерь

**Лічильник** — накопичувач одиниць: перед циклом `0`, і `+= 1` щоразу, коли виконується умова.


````python
dinners = 0
i = 0
while i < len(orders):
    if orders[i].time == "вечеря":
        dinners += 1
    i += 1

print("Вечерь:", dinners, "з", len(orders), "чеків")
````

### Алгоритм 4 — фільтр у новий список: великі столи

**Фільтр** — накопичувач елементів: перед циклом порожній список, і `.append()` для кожного елемента, що пройшов умову.

> Правило: **ніколи не видаляй елементи зі списку, по якому зараз ідеш** — індекси зсуваються, і частина елементів пропускається. Збирай результат в окремий список.


````python
large_tables = []
i = 0
while i < len(orders):
    if orders[i].size >= 4:
        large_tables.append(orders[i])
    i += 1

print("Великих столів:", len(large_tables))
print("orders не змінився:", len(orders), "чеків")
````

### Пройти відфільтрований список

Результат фільтра — звичайний список `Order`. Його проходимо тим самим `while` і для кожного великого стола рахуємо чек на одну людину.

```text
   Три форми одного циклу:    накопичувач   total += order.total_bill
                              лічильник     count += 1
                              фільтр        new_list.append(order)
```


````python
j = 0
while j < len(large_tables):
    o = large_tables[j]
    print(f"  {o.waiter:<8} {o.total_bill:>6} грн на {o.size} гостей = {o.total_bill / o.size:.1f} грн/особу")
    j += 1
````

## 5. Множина — адміністратор і гості

**Проблема.** Власниця питає: «У які дні ми працювали?» У списку чеків день повторюється в кожному чеку. «Хто з офіціантів виходив?» — те саме. «Хто з гостей був і в п'ятницю, і в суботу?» — треба порівняти два набори імен. У всіх трьох питаннях **важлива лише наявність**, не порядок і не кількість повторів.

**Дані.** Значення, серед яких багато дублікатів, а потрібні лише унікальні.

**Тип.** `set` — набір унікальних значень без порядку.


### Створити множину — дублікати зникають самі

Множину записують у **фігурних дужках**. Однакові значення в ній зберігаються лише один раз. Порядок, у якому Python її друкує, не гарантований.


````python
tables = {3, 1, 2, 3, 1}
print(tables)
print("унікальних столів:", len(tables))
````

### 🔮 Порожня множина — тільки `set()`

Що таке `{}`? Це **порожній словник** (тема уроку 6), а не множина. Порожню множину створює лише `set()`.


````python
print("type({})    =", type({}).__name__)
print("type(set()) =", type(set()).__name__)
````

### `.add(x)` — додати елемент

`.add()` додає одне значення. Якщо воно вже є — нічого не відбувається і помилки немає. Саме тому множина сама прибирає дублікати.


````python
days = set()
days.add("пт")
days.add("сб")
days.add("пт")                    # уже є — ігнорується
print(days)
````

### `in` — швидка перевірка наявності

Перевірка належності — головна суперсила множини. `in` для множини працює миттєво навіть на мільйонах елементів, а список переглядається по одному.


````python
print("працювали в суботу?", "сб" in days)
print("працювали в понеділок?", "пн" in days)
````

### `.discard(x)` — видалити, якщо є

`.discard()` прибирає значення, а якщо його немає — тихо нічого не робить. Зручно, коли не знаєш напевно, чи елемент є.


````python
days.discard("пн")                # немає — без помилки
days.discard("сб")
print(days)
days.add("сб")                    # повертаємо для наступних кроків
````

### `.remove(x)` — видалити або помилка

`.remove()` у множини поводиться суворіше: якщо значення немає — `KeyError`. Обирай його, коли відсутність елемента означає баг, який треба помітити.


````python
try:
    days.remove("пн")
except KeyError as e:
    print("KeyError:", e)
````

### 🔮 Індекс у множини

Що дасть `days[0]`? У множини немає «першого» елемента, тож і індексів немає — `TypeError`.


````python
try:
    days[0]
except TypeError as e:
    print("TypeError:", e)
````

### `sorted()` — впорядкований вивід

Коли порядок потрібен для показу, `sorted(множина)` повертає **відсортований список**. Сама множина від цього не змінюється.


````python
print(sorted(days))
print(type(sorted(days)).__name__)
````

### Унікальні дні та офіціанти з чеків

Той самий прохід `while`, але замість накопичувача — множина. Повторні `"пт"` або `"Тарас"` вона просто не додає.


````python
days = set()
waiters = set()
i = 0
while i < len(orders):
    days.add(orders[i].day)
    waiters.add(orders[i].waiter)
    i += 1

print(f"{len(orders)} чеків, але днів лише {len(days)}:", sorted(days))
print("офіціантів на зміні:", len(waiters), sorted(waiters))
````

### `set(список)` — унікальні значення одним рядком

Для готового списку значень те саме робить `set(...)`. Пам'ятай: порядок після цього губиться.

```text
   ["кава", "чай", "кава"]  →  {"кава", "чай"}  →  sorted → ["кава", "чай"]
```


````python
dishes = ["кава", "чай", "кава", "узвар", "чай"]
print("унікальних страв:", len(set(dishes)), sorted(set(dishes)))
````

### Гості двох днів

Адміністраторка записувала гостей. Кожне ім'я важливе лише один раз, тому гості кожного дня — множина. Далі порівнюємо два набори.

```text
       friday                 saturday
   ┌─────────────┬───────────┬─────────────┐
   │ Олена       │  Марта    │             │
   │ Богдан      │  Ігор     │  Софія      │
   └─────────────┴───────────┴─────────────┘
      f - s          f & s        s - f
```


````python
friday   = {"Олена", "Богдан", "Марта", "Ігор"}
saturday = {"Марта", "Ігор", "Софія"}
print("пт:", sorted(friday))
print("сб:", sorted(saturday))
````

### `|` — об'єднання: хто приходив хоча б раз

`a | b` (або `a.union(b)`) — усі елементи обох множин, кожен один раз.


````python
print(sorted(friday | saturday))
````

### `&` — перетин: хто був обидва дні

`a & b` (або `a.intersection(b)`) — лише спільні елементи. Постійні гості, яким кафе дарує десерт.


````python
print(sorted(friday & saturday))
````

### `-` — різниця: був у п'ятницю, але не в суботу

`a - b` (або `a.difference(b)`) — елементи `a`, яких немає в `b`. Порядок важливий: `b - a` дасть інший результат.


````python
print("пт - сб:", sorted(friday - saturday))
print("сб - пт:", sorted(saturday - friday))
````

### `^` — симетрична різниця: лише в один із днів

`a ^ b` (або `a.symmetric_difference(b)`) — елементи, що є рівно в одній із множин.


````python
print(sorted(friday ^ saturday))
````

### `.issubset()` — чи всі елементи є в іншій множині

`a.issubset(b)` повертає `True`, якщо кожен елемент `a` є в `b`. Наприклад: чи всі постійні гості були в п'ятницю?


````python
regulars = friday & saturday
print("постійні ⊂ п'ятниця?", regulars.issubset(friday))
print("субота ⊂ п'ятниця?  ", saturday.issubset(friday))
````

### Оператор чи метод?

Оператори `| & - ^` працюють **лише між двома множинами**. Методи з назвами (`.union()`, `.intersection()` …) приймають і список.


````python
print(sorted(friday.union(["Зоя"])))
try:
    friday | ["Зоя"]
except TypeError as e:
    print("TypeError:", e)
````

### Що можна класти в множину

Елементи множини мають бути **незмінними**: числа, рядки, кортежі. Список покласти не можна — `TypeError: unhashable type`. Тому пару «стіл, офіціант» зберігають кортежем.


````python
served = {(4, "Тарас"), (2, "Марія"), (4, "Тарас")}
print(served)
try:
    {[4, "Тарас"]}
except TypeError as e:
    print("TypeError:", e)
````

### `frozenset` — незмінна множина

`frozenset(...)` — множина, яку не можна змінити. Зате її можна покласти в іншу множину або використати там, де потрібне незмінне значення.

```text
                SET — унікальні, без порядку, без індексів
  ┌───────────────┬─────────────────┬────────────────────┬───────────────────────────┐
  │ створити      │ змінити         │ перевірити         │ порівняти набори          │
  ├───────────────┼─────────────────┼────────────────────┼───────────────────────────┤
  │ {1, 2}        │ .add(x)         │ x in s             │ a | b   .union()          │
  │ set()  (порож)│ .discard(x)     │ len(s)             │ a & b   .intersection()   │
  │ set(list)     │ .remove(x) !Key │ sorted(s) → list   │ a - b   .difference()     │
  │ frozenset(..) │ .pop() .clear() │                    │ a ^ b   .symmetric_diff.. │
  │               │                 │                    │ .issubset() .isdisjoint() │
  └───────────────┴─────────────────┴────────────────────┴───────────────────────────┘
```


````python
shift = frozenset({"Тарас", "Марія"})
print(shift)
try:
    shift.add("Олексій")
except AttributeError as e:
    print("AttributeError:", e)
````

## 6. Звіт за день — усі три типи разом

Збираємо все в одну програму. **Один прохід** списком оновлює одразу накопичувачі, чемпіона і дві множини: `list` тримає чеки, `Order` (кортеж) — поля кожного чека, `set` — унікальні дні й офіціантів.

```text
  orders: list[Order]
     │  while i < len(orders):  order = orders[i]
     ├── revenue += order.total_bill ─┐
     ├── tips    += order.tip         ├─► числа (накопичувачі)
     ├── best = order, якщо більше   ─┘  (поточний чемпіон)
     ├── days.add(order.day)         ─┐
     └── waiters.add(order.waiter)   ─┴─► множини (унікальні)
                                              ▼
                                         print(звіт)
```


````python
revenue = 0
tips = 0
best = orders[0]
days = set()
waiters = set()

i = 0
while i < len(orders):
    order = orders[i]
    revenue += order.total_bill
    tips += order.tip
    if order.total_bill > best.total_bill:
        best = order
    days.add(order.day)
    waiters.add(order.waiter)
    i += 1

print("=" * 40)
print("        ЗВІТ ЗА ЗМІНУ")
print("=" * 40)
print("Чеків:          ", len(orders))
print("Виторг:         ", revenue, "грн")
print("Чайові:         ", tips, "грн  (", round(tips / revenue * 100, 1), "% )")
print("Найбільший чек: ", best.total_bill, "грн —", best.waiter)
print("Днів роботи:    ", len(days), sorted(days))
print("Офіціантів:     ", len(waiters), sorted(waiters))
print("=" * 40)
````

## 7. Питання, на яке ми ще не вміємо відповісти

Власниця задоволена звітом і питає останнє: **«А скільки чайових заробив кожен офіціант окремо?»**

Подивимось, що в нас є:

- `tips = 270.0` — одне число на всіх. Накопичувач не знає, чиї це гроші.
- `waiters = {'Марія', 'Олексій', 'Тарас'}` — множина знає **хто**, але не вміє тримати число поруч з іменем.
- Можна завести три змінні `tips_taras`, `tips_maria`, `tips_oleksii`… а завтра вийде четвертий офіціант, і програму доведеться переписувати.

Потрібна структура, яка **зв'язує ім'я з числом**: «Тарас → 170.0, Марія → 30.0, Олексій → 70.0». Це **словник** — тема уроку 6. Там цей самий `orders` і цей самий звіт отримають рядок «чайові по офіціантах», а цикл `while` з індексом стане коротшим завдяки `for`.


## 8. Формула уроку

```text
ДАНІ  →  СТРУКТУРА  →  ЦИКЛ + УМОВА  →  НАКОПИЧЕННЯ  →  РЕЗУЛЬТАТ
чеки     list[Order]    while / if       += , best, set.add    print(звіт)
```

| Задача в кафе | Контейнер | Чим користуємось |
|---|---|---|
| замовлення столика росте й змінюється | `list` | `.append .insert .remove .pop .extend`, індекси, зрізи |
| закритий чек не можна змінити | `tuple` | розпакування `table, total, tip = receipt` |
| чек із шістьма полями, що мають сенс | `NamedTuple` | `order.tip`, `._replace()`, досі `tuple` |
| усі чеки дня | `list[Order]` | `while i < len(orders)`, накопичувач, «поточний чемпіон», фільтр у новий список |
| дні роботи, офіціанти, гості | `set` | `.add`, `in`, `| & - ^` |
| «ім'я → число» | — | словник, урок 6 |

Вибір контейнера — це повідомлення читачеві коду: «цей набір росте» (`list`), «цей запис не зміниться» (`tuple`), «тут важлива лише наявність» (`set`).
