# ☕ Вечірня зміна: чи отримає Тарас чайові?

**Урок 5 · Списки, кортежі та множини · квест для груп**

Кафе зачиняється. Офіціант Тарас цілий вечір бігав між столами, і тепер чекає на чайові. Але власниця кафе видасть їх лише тоді, коли **каса зійдеться**: чотири служби мають перевірити свою частину дня — і кожна працює зі своїм контейнером.

```text
                 💰 чайові Тараса
                        ↑
               🧾 закриття зміни
     ┌──────────┬───────┴────┬──────────────┐
  1. Кухня   2. Каса      3. Зал     4. Адміністратор
   list       tuple    NamedTuple         set
             розпакування  індекси      перетин
```

Якщо хоча б одна перевірка не зійдеться — Тарас піде додому без чайових.

3. Якщо все правильно, каса покаже **частину чайових** у гривнях. Назвіть цю суму викладачу.
4. Викладач вписує чотири суми в клітинку «🧾 Закриття зміни» — і всі дізнаються, скільки отримає Тарас.

**Самостійно (удома).** Пройди всі чотири перевірки. Суми збережуться у змінних `TIP_1` … `TIP_4`, і зміну можна закрити одним рядком.

> Каса не зберігає правильних відповідей відкритим текстом, тож підглянути не вийде. Але на типову помилку вона дасть підказку.

````python
# @title 🔒 Каса кафе — натисніть ▶ один раз
# Не змінюйте цю клітинку: тут перевірки й підсумок зміни.
import hashlib

_SALT = "cafe-shift-l5"
_ANSWERS = {1: "403769ffc3c45751f3af3d2f0401718b7001b5b1586c6c4cc2b3304db0d8c59b", 2: "589aeb4375a62fa1e8f243efb2e43783f146cf7842800af55d493f960f717e76", 3: "25b423be85354bbbe421db77dc0006dfadf4f080be7aa92c59ea2e7d9fa891b1", 4: "dc23cd458a03c73cbdc3675f54543ae583ad834b087070940b7bd2c031053673"}
_TIPS = {1: "f901e39b8303666386b7b1fbcb3b02a421626b4b689a6e1957465fe82e5f76a3", 2: "f27c56d502d7bf612150af26af60595ccc4ee92af75e38261c7358d05f6a09a1", 3: "e84fa3675869e799ee40ca7a8ff1f4a8a9ff9e898909e25bcba6c3a63e4808e7", 4: "12ae23d6b226d8ebf709f74101ea851b89569840857778159809a9edc08e602a"}
_CHECKS = {1: "Кухня", 2: "Каса", 3: "Зал", 4: "Адміністратор"}


def _digest(text):
    return hashlib.sha256((_SALT + text).encode("utf-8")).hexdigest()


def _normal(answer):
    if isinstance(answer, bool):
        return "bool:" + repr(answer)
    if isinstance(answer, (int, float)):
        return "number:" + repr(float(answer))
    if isinstance(answer, (set, frozenset)):
        return "set:" + repr(sorted(answer, key=repr))
    return type(answer).__name__ + ":" + repr(answer)


def _tip(n, answer):
    return int(_digest("tip" + str(n) + _normal(answer))[:8], 16) % 91 + 40


def _hint(n, answer):
    if n == 1:
        if isinstance(answer, tuple):
            return "Замовлення змінюється, тому це має бути список, а не кортеж."
        if not isinstance(answer, list):
            return "Відповідь — сам список позицій items після всіх змін."
        if "кава" in answer:
            return "Гість відмовився від кави — прибери її методом .remove()."
        if "узвар" not in answer:
            return "Гість дозамовив узвар — додай його в кінець методом .append()."
        if "хліб" in answer and answer[0] != "хліб":
            return "Хліб кухня подає першим — він має стояти на початку списку: .insert(0, ...)."
        return "Перевір порядок змін: спершу додали узвар, потім прибрали каву, потім хліб на початок."
    if n == 2:
        if isinstance(answer, (int, float)) and answer == 860:
            return "Це сума замовлення без чайових. До сплати — замовлення плюс чайові."
        if isinstance(answer, tuple):
            return "Розпакуй чек у три змінні: table, total, tip = receipt."
        return "До сплати = сума замовлення + чайові. Номер столу в розрахунок не входить."
    if n == 3:
        if isinstance(answer, (int, float)) and answer == 200:
            return "Це чайові з перших трьох чеків. Потрібні останні три: індекси -3, -2, -1."
        if isinstance(answer, list):
            return "Потрібна сума чайових, одне число, а не список чеків."
        return "Візьми останні три чеки через від'ємні індекси і додай їхні поля .tip."
    if n == 4:
        if isinstance(answer, list):
            return "Потрібна множина (set), а не список."
        if isinstance(answer, set) and len(answer) == 5:
            return "Це всі гості двох днів (|). Постійні — ті, хто був і в п'ятницю, і в суботу: &."
        if isinstance(answer, set) and "Олена" in answer:
            return "Олена була лише в п'ятницю. Потрібні гості, що були в обидва дні."
        return "Постійні гості — перетин двох множин: friday & saturday."
    return ""


def check(n, answer):
    """Перевіряє відповідь групи n. Якщо все правильно — повертає частину чайових (грн)."""
    name = _CHECKS[n]
    if answer is None:
        print(f"✏️ {name}: впишіть відповідь замість None і запустіть клітинку ще раз.")
        return None
    if _digest("answer" + _normal(answer)) == _ANSWERS[n]:
        tip = _tip(n, answer)
        print(f"✅ {name}: сходиться!")
        print(f"💵 Частина чайових від перевірки {n}: {tip} грн")
        print("   Назвіть цю суму викладачу.")
        return tip
    print(f"❌ {name}: не сходиться.")
    print("💡 " + _hint(n, answer))
    return None


def close_shift(tip_1=None, tip_2=None, tip_3=None, tip_4=None):
    """Закриття зміни: офіціант отримує чайові, лише якщо всі чотири перевірки зійшлися."""
    tips = {1: tip_1, 2: tip_2, 3: tip_3, 4: tip_4}
    print("🧾 Закриваємо зміну, звіряємо касу...\n")
    wrong = []
    for n, tip in tips.items():
        if tip is not None and _digest("tip-check" + str(n) + str(tip)) == _TIPS[n]:
            print(f"   ✅ {_CHECKS[n]}: {tip} грн")
        else:
            print(f"   ❌ {_CHECKS[n]}: суми немає або вона не та")
            wrong.append(n)
    print()
    if wrong:
        print("😞 Каса не зійшлася — сьогодні Тарас без чайових.")
        print("   Перевірте:", ", ".join(_CHECKS[n] for n in wrong))
        return None
    total = tip_1 + tip_2 + tip_3 + tip_4
    print("🎉 Каса зійшлася!")
    print(f"💰 Офіціант Тарас отримує чайові: {total} грн")
    print()
    print("Тарас дякує команді:")
    print("   «Кухня вела список замовлення, каса розпакувала чек,")
    print("    зал знайшов останні чеки за індексами, а адміністратор")
    print("    перетином множин знайшов постійних гостей.")
    print("    Список, кортеж, множина — кожен на своєму місці!»")
    return total


print("🔧 Каса кафе готова ✅")
````
