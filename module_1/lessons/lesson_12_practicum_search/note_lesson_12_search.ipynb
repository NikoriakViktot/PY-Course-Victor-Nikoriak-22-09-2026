## 1. Чи їздив клієнт сьогодні? — лінійний пошук

Журнал поїздок записаний у порядку дзвінків, імена йдуть як завгодно. Скористатися нічим: перевіряємо кожен запис. Функція повертає пару `(індекс, кроки)`; `-1` — не знайдено.

```python
def linear_search(items, target):
    steps = 0
    for i in range(len(items)):
        steps += 1
        if items[i] == target:
            return i, steps
    return -1, steps
```

```python
clients = ["Марта", "Олег", "Ірина", "Тарас", "Олена", "Богдан"]
print(linear_search(clients, "Тарас"))
print(linear_search(clients, "Софія"))

assert linear_search(clients, "Тарас") == (3, 4)
assert linear_search(clients, "Софія") == (-1, 6)   # немає — перевірили всіх
assert linear_search([], "Тарас") == (-1, 0)
```

## 2. Чи була поїздка о 18:40? — бінарний пошук

Система записує час початку поїздок у хвилинах від опівночі (18:40 → `18 * 60 + 40 = 1120`). Записи додаються по порядку, тож список **відсортований**.

Дивимось на середній запис: якщо там 15:30, а шукаємо 18:40, уся ліва половина ще раніша — її можна відкинути. Одне порівняння — мінус половина даних.

**Передбач:** скільки кроків знадобиться на 1000 записах? На 2000? На мільйоні?

```python
def binary_search(items, target):
    low, high = 0, len(items) - 1
    steps = 0
    while low <= high:
        mid = (low + high) // 2
        steps += 1
        if items[mid] == target:
            return mid, steps
        if items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, steps
```

```python


starts = [425, 510, 612, 700, 845, 930, 1035, 1120, 1210, 1290, 1375]
print(binary_search(starts, 1120))
print(binary_search(starts, 1000))

assert binary_search(starts, 1120) == (7, 4)
assert binary_search(starts, 1000)[0] == -1
```

Як шукалося 1120 — пройди таблицю пальцем по списку `starts`:

| Крок | `low` … `high` | `mid` → значення | Рішення |
|---|---|---|---|
| 1 | 0 … 10 | 5 → 930 | менше → `low = 6` |
| 2 | 6 … 10 | 8 → 1210 | більше → `high = 7` |
| 3 | 6 … 7 | 6 → 1035 | менше → `low = 7` |
| 4 | 7 … 7 | 7 → 1120 | знайдено |

Тепер дослід подвоєння. Шукаємо значення, якого немає (найгірший випадок):

```python
for n in [1000, 2000, 4000, 1_000_000]:
    times = list(range(0, 2 * n, 2))
    _, steps = binary_search(times, -1)
    print(f"{n:>9} записів → {steps} кроків")
```

Подвоїли дані — **+1 крок**. Мільйон записів — 19 кроків (лінійному пошуку — мільйон). Кроків стільки, скільки разів n ділиться навпіл до одиниці: `log₂ n`. Це клас `O(log n)`.

⚠️ **Лише для відсортованих даних.** На невідсортованому списку пошук не падає — він мовчки бреше:

```python
print(binary_search([1120, 425, 930, 612], 1120))   # 1120 є в списку, але пошук її не знайде
```

## 3. Ваучер на 500 грн — два вказівники

Пасажир отримав ваучер на **рівно 500 грн** і може покрити ним дві поїздки. Суми поїздок за тиждень **відсортовані**. Які дві поїздки дають рівно 500?

Перебір усіх пар — `O(n²)`. Два вказівники — з обох кінців назустріч:
- сума **замала** → лівий вказівник праворуч (до дорожчої поїздки);
- сума **завелика** → правий ліворуч (до дешевшої);
- рівно 500 → знайшли.

**Завдання.** Допиши тіло циклу. Функція повертає `(пара, кроки)` або `(None, кроки)`.

```python
fares = [120, 150, 180, 230, 270, 320, 410]


def pair_with_sum(items, target):
    left, right = 0, len(items) - 1
    steps = 0
    while left < right:
        steps += 1
        # YOUR CODE HERE
        # BEGIN SOLUTION
        total = items[left] + items[right]
        if total == target:
            return (items[left], items[right]), steps
        if total < target:
            left += 1
        else:
            right -= 1
        # END SOLUTION
    return None, steps


def pair_with_sum_brute(items, target):
    steps = 0
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            steps += 1
            if items[i] + items[j] == target:
                return (items[i], items[j]), steps
    return None, steps


print("два вказівники:", pair_with_sum(fares, 500))
print("перебір пар:   ", pair_with_sum_brute(fares, 500))
print("пари немає:    ", pair_with_sum(fares, 10_000), "проти", pair_with_sum_brute(fares, 10_000))

assert pair_with_sum(fares, 500) == ((180, 320), 4)
assert pair_with_sum(fares, 10_000) == (None, 6)
assert pair_with_sum([200, 300], 500) == ((200, 300), 1)
assert pair_with_sum([], 500) == (None, 0)
print("OK")
```

Чому можна відкинути 410 на першому кроці? `120 + 410 = 530` — забагато навіть з **найдешевшою** поїздкою, а з дорожчими буде ще більше. Кожен крок відкидає одну поїздку назавжди: щонайбільше n − 1 кроків, `O(n)`.

## 4. Пікові години — ковзне вікно

Кількість поїздок по годинах доби (індекс — година). Коли почати пікову зміну водіїв: які **три години поспіль** найзавантаженіші?

Наївний спосіб рахує суму кожного вікна заново — `O(n·k)`. Але сусідні вікна відрізняються двома годинами: одна випадає зліва, одна додається справа.

**Завдання.** Допиши `busiest_window`: оновлюй суму вікна, а не рахуй її заново. Результат має збігатися з наївною версією для будь-якого k.

```python
per_hour = [3, 1, 0, 0, 1, 4, 12, 30, 41, 25, 18, 20,
            22, 19, 17, 21, 33, 45, 52, 38, 24, 15, 9, 6]


def busiest_window_naive(counts, k):
    best_start, best_sum = 0, sum(counts[0:k])
    for start in range(1, len(counts) - k + 1):
        window_sum = sum(counts[start:start + k])
        if window_sum > best_sum:
            best_start, best_sum = start, window_sum
    return best_start, best_sum


def busiest_window(counts, k):
    window_sum = sum(counts[0:k])
    best_start, best_sum = 0, window_sum
    for end in range(k, len(counts)):
        # YOUR CODE HERE
        # BEGIN SOLUTION
        window_sum += counts[end] - counts[end - k]
        if window_sum > best_sum:
            best_start, best_sum = end - k + 1, window_sum
        # END SOLUTION
    return best_start, best_sum


print(busiest_window(per_hour, 3))
assert busiest_window(per_hour, 3) == (17, 135)      # 17:00–20:00, 135 поїздок
for k in range(1, 25):
    assert busiest_window(per_hour, k) == busiest_window_naive(per_hour, k), k
print("OK — збігається з наївною версією для k від 1 до 24")
```

## 5. Автодоповнення адреси

Пасажир вводить вулицю, і після кожної літери з'являються підказки. Вулиці відсортовані за абеткою, тож усі, що починаються з префікса, стоять **підряд** — з позиції, яку дає `bisect_left`.

**Завдання.** Допиши `suggest`: знайди початок через `bisect_left`, далі йди праворуч, поки вулиця починається з префікса (`str.startswith`) і підказок менше за `limit`.

```python
streets = sorted(["Богдана Хмельницького", "Велика Васильківська", "Верхній Вал",
                  "Володимирська", "Хрещатик", "Хорива", "Шота Руставелі"])


def suggest(streets, prefix, limit=3):
    result = []
    # YOUR CODE HERE
    # BEGIN SOLUTION
    i = bisect_left(streets, prefix)
    while i < len(streets) and streets[i].startswith(prefix) and len(result) < limit:
        result.append(streets[i])
        i += 1
    # END SOLUTION
    return result


print(suggest(streets, "В"))
assert suggest(streets, "В") == ["Велика Васильківська", "Верхній Вал", "Володимирська"]
assert suggest(streets, "Ве") == ["Велика Васильківська", "Верхній Вал"]
assert suggest(streets, "Х") == ["Хорива", "Хрещатик"]
assert suggest(streets, "Я") == []
assert suggest(streets, "В", limit=1) == ["Велика Васильківська"]
print("OK")
```

## 6. Готові інструменти: Python і бібліотеки

Свої пошуки ми писали, щоб **зрозуміти** ідею: чому бінарний пошук робить 20 кроків замість мільйона. У робочому коді те саме зазвичай беруть готовим:

| Задача | Наш код | Стандартна бібліотека | Сторонні бібліотеки |
|---|---|---|---|
| чи є значення | `linear_search` | `x in list` | — |
| пошук у відсортованому | `binary_search` | `bisect.bisect_left` | `numpy.searchsorted`, `sortedcontainers.SortedList` |
| сума вікна | `busiest_window` | `itertools.accumulate` | `numpy.cumsum`, `pandas`: `rolling(k).sum()` |
| автодоповнення | `suggest` | `bisect` + `str.startswith` | `SortedList.irange`, `pandas`: `str.startswith` |

Готові інструменти написані на C, тому швидші за наш цикл на Python. Чи скасовує це Big O?

Час міряємо модулем `timeit`: `Timer(...).autorange()` сам підбирає, скільки разів запустити функцію, щоб замір був точним.

**Передбач:** хто швидший на мільйоні записів — `x in list` (на C, але `O(n)`) чи наш `binary_search` (на Python, але `O(log n)`)?

```python
import timeit
from bisect import bisect_left, bisect_right


def per_call(func):
    """Середній час одного виклику func(), у мікросекундах."""
    number, total = timeit.Timer(func).autorange()
    return total / number * 1_000_000


journal = list(range(0, 2_000_000, 2))   # мільйон відсортованих моментів
missing = -1                              # такого моменту немає — найгірший випадок

print(f"наш linear_search: {per_call(lambda: linear_search(journal, missing)):>10.1f} мкс")
print(f"x in list:         {per_call(lambda: missing in journal):>10.1f} мкс")
print(f"наш binary_search: {per_call(lambda: binary_search(journal, missing)):>10.1f} мкс")
print(f"bisect_left:       {per_call(lambda: bisect_left(journal, missing)):>10.1f} мкс")
```

`x in list` у кілька разів швидший за наш цикл: це той самий лінійний пошук, але на C. Проте він усе одно перевіряє мільйон записів, тому **наш** бінарний пошук на Python обганяє його в сотні разів. А `bisect_left` і написаний на C, і має `O(log n)`, тому він найшвидший.

**Правило:** спершу обери правильний алгоритм (Big O), потім — швидку реалізацію.

### Родина `bisect`

- `bisect_left(a, x)` — перша позиція, куди можна вставити `x` (перед рівними);
- `bisect_right(a, x)` — позиція після всіх рівних;
- різниця двох позицій — скільки значень у проміжку;
- `insort(a, x)` — вставити, зберігши порядок (місце шукається за `O(log n)`, але зсув елементів — `O(n)`);
- з Python 3.10 є параметр `key=` — шукати в списку кортежів за одним полем.

```python
from bisect import insort

# Скільки поїздок почалося з 18:00 до 22:00 (не включно)?
low = bisect_left(starts, 18 * 60)
high = bisect_left(starts, 22 * 60)
print("позиції:", low, high, "→ поїздок:", high - low, starts[low:high])

# Скільки поїздок почалося до 18:40 включно?
print("до 18:40 включно:", bisect_right(starts, 1120))

# Нова поїздка о 17:00 — вставляємо, не ламаючи порядку
evening_starts = starts.copy()
insort(evening_starts, 17 * 60)
print("після insort:", evening_starts)

# Поїздки — кортежі (час, клієнт), відсортовані за часом; шукаємо за полем «час»
trips = [(425, "Марта"), (612, "Олег"), (930, "Ірина"), (1120, "Тарас"), (1290, "Олена")]
i = bisect_left(trips, 1120, key=lambda trip: trip[0])
print("о 18:40 поїхав:", trips[i][1])

assert high - low == 3 and bisect_right(starts, 1120) == 8
assert evening_starts == sorted(evening_starts) and 1020 in evening_starts
assert trips[i] == (1120, "Тарас")
```

### numpy: один запит чи тисячі

`numpy.searchsorted(масив, x)` — той самий `bisect_left`, але для масиву numpy. Головне вміння numpy — **векторизація**: передати не одне значення, а цілий масив запитів, і numpy обробить їх усі всередині C, без циклу Python.

**Передбач:** для одного запиту numpy швидший за `bisect`? А для 100 000 запитів?

```python
import random

import numpy as np

random.seed(12)
array = np.asarray(journal)               # перетворення списку на масив — один раз, теж коштує часу
print(f"один запит, bisect_left:     {per_call(lambda: bisect_left(journal, 1120)):8.2f} мкс")
print(f"один запит, np.searchsorted: {per_call(lambda: np.searchsorted(array, 1120)):8.2f} мкс")

queries = [random.randrange(2_000_000) for _ in range(100_000)]
query_array = np.asarray(queries)
loop_us = per_call(lambda: [bisect_left(journal, q) for q in queries])
vector_us = per_call(lambda: np.searchsorted(array, query_array))
print(f"100 000 запитів, цикл + bisect:        {loop_us / 1000:8.1f} мс")
print(f"100 000 запитів, numpy одним викликом: {vector_us / 1000:8.1f} мс")

assert list(np.searchsorted(array, query_array[:1000])) == [bisect_left(journal, q) for q in queries[:1000]]
```

Для **одного** запиту numpy повільніший: кожен виклик numpy має накладні витрати. Для тисяч запитів одним викликом — швидший, бо цикл іде на C.

### Ковзне вікно — готовим

У ковзного вікна є готова форма — **префіксні суми**. `prefix[i]` — сума перших `i` годин; сума вікна від `s` довжини `k` — це `prefix[s + k] - prefix[s]`, одна дія. Префіксні суми рахують `itertools.accumulate` і `numpy.cumsum`, а `pandas` має готове `rolling(k).sum()`.

```python
from itertools import accumulate

import pandas as pd

k = 3
prefix = [0, *accumulate(per_hour)]
sums = [prefix[s + k] - prefix[s] for s in range(len(per_hour) - k + 1)]
best = max(range(len(sums)), key=lambda s: sums[s])
print("itertools.accumulate:", best, sums[best])

np_prefix = np.concatenate(([0], np.cumsum(per_hour)))
np_sums = np_prefix[k:] - np_prefix[:-k]
print("numpy.cumsum:        ", int(np.argmax(np_sums)), int(np_sums.max()))

rolling = pd.Series(per_hour).rolling(k).sum()      # сума вікна, що ЗАКІНЧУЄТЬСЯ на цьому індексі
end = int(rolling.idxmax())
print("pandas rolling:      ", end - k + 1, int(rolling[end]))

assert (best, sums[best]) == busiest_window_naive(per_hour, k) == (17, 135)
assert int(np.argmax(np_sums)) == 17 and end - k + 1 == 17
```

### Автодоповнення: `SortedList` і пастка «швидкої бібліотеки»

`sortedcontainers.SortedList` — список, що **завжди** відсортований: `add` ставить новий елемент на місце, а `in` та `irange` шукають бінарно. Зручно, коли довідник постійно поповнюється. Якщо бібліотеки немає: `!pip install sortedcontainers`.

`pandas` уміє `str.startswith` для цілого стовпця одразу, на C.

**Передбач:** хто швидший на 300 000 адрес — `pandas` чи `bisect`?

```python
from itertools import islice

from sortedcontainers import SortedList

random.seed(12)
stems = ["Антоновича", "Басейна", "Велика Васильківська", "Володимирська", "Хорива",
         "Хрещатик", "Шота Руставелі", "Ярославів Вал"]
book = sorted(f"{random.choice(stems)}, {i}" for i in range(300_000))
book_sorted = SortedList(book)
book_series = pd.Series(book)
prefix = "Хр"


def by_loop():
    return [s for s in book if s.startswith(prefix)][:3]


def by_pandas():
    return book_series[book_series.str.startswith(prefix)].head(3).tolist()


def by_bisect():
    i = bisect_left(book, prefix)
    return [s for s in book[i:i + 3] if s.startswith(prefix)]


def by_sorted_list():
    # irange віддає елементи між двома межами; "￿" більше за будь-яку літеру
    return list(islice(book_sorted.irange(prefix, prefix + "￿"), 3))


for name, func in [("цикл + startswith", by_loop), ("pandas str.startswith", by_pandas),
                   ("bisect", by_bisect), ("SortedList.irange", by_sorted_list)]:
    print(f"{name:<22} {per_call(func):>10.1f} мкс  {func()}")

assert by_loop() == by_pandas() == by_bisect() == by_sorted_list()

book_sorted.add("Хрещатик, 0")            # нова адреса одразу стає на своє місце
print("після add:", by_sorted_list())
```

Що показав розділ:

- готові інструменти на C швидші за наш код на Python у рази;
- але `x in list` і `pandas str.startswith` усе одно перевіряють **кожен** запис — `O(n)`. Їх обганяє `bisect`, бо `O(log n)` на мільйоні — це 20 кроків;
- `numpy` виграє на **багатьох** запитах одним викликом, а не на одному;
- перетворення даних (`np.asarray`, `SortedList(...)`, `pd.Series`) коштує часу — воно окупається, коли запитів багато.

**Правило:** алгоритм визначає, як росте робота, бібліотека — наскільки швидкий кожен крок. Спершу Big O, потім бібліотека.

### Тихі години для оновлення сервера

Потрібно **k годин поспіль** з найменшою кількістю поїздок. Напиши `quietest_window(counts, k)` — ковзне вікно, що шукає мінімум. При однакових сумах повертай найраніше вікно.

```python
def quietest_window(counts, k):
    # YOUR CODE HERE
    # BEGIN SOLUTION
    window_sum = sum(counts[0:k])
    best_start, best_sum = 0, window_sum
    for end in range(k, len(counts)):
        window_sum += counts[end] - counts[end - k]
        if window_sum < best_sum:
            best_start, best_sum = end - k + 1, window_sum
    return best_start, best_sum
    # END SOLUTION


assert quietest_window(per_hour, 2) == (2, 0)
assert quietest_window(per_hour, 4) == (1, 2)
print("OK")
```

### Перша поїздка з певного часу — власний `bisect_left`

Напиши бінарний пошук `first_at_or_after(items, target)` **без** модуля `bisect`: індекс першого елемента, не меншого за `target`, або `len(items)`, якщо таких немає. Порахуй порівняння: їх має бути не більше `len(items).bit_length() + 1`.

```python
def first_at_or_after(items, target):
    """Повертає (індекс, кроки)."""
    steps = 0
    # YOUR CODE HERE
    # BEGIN SOLUTION
    low, high = 0, len(items)
    while low < high:
        mid = (low + high) // 2
        steps += 1
        if items[mid] < target:
            low = mid + 1
        else:
            high = mid
    return low, steps
    # END SOLUTION


assert first_at_or_after(starts, 1080)[0] == 7
assert first_at_or_after(starts, 1400)[0] == 11
assert first_at_or_after([], 5)[0] == 0
for t in range(400, 1401):
    index, steps = first_at_or_after(starts, t)
    assert index == bisect_left(starts, t), t
    assert steps <= len(starts).bit_length() + 1
print("OK — збігається з bisect_left для всіх часів від 400 до 1400")
```
