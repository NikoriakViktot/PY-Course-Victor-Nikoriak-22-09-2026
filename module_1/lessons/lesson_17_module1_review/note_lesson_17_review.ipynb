### Прогноз 1. Урок 3 — змінна як ярлик

```python
a = [540, 320]
b = a
b.append(980)
print(a)
```

### Прогноз 3. Урок 6 — лічильник у словнику

```python
counts = {}
for day in ["пт", "сб", "пт"]:
    counts[day] = counts.get(day, 0) + 1
print(counts)
```

<details>
<summary>Відповідь</summary>

`{'пт': 2, 'сб': 1}`.

</details>

### Прогноз 4. Урок 7 — значення за замовчуванням

```python
def add_order(bill, orders=[]):
    orders.append(bill)
    return orders


add_order(540)
print(add_order(320))
```

<details>
<summary>Відповідь</summary>

`[540, 320]` — список за замовчуванням створюється **один раз**, при оголошенні функції. Правильно: `orders=None` і `if orders is None: orders = []`.

</details>

### Прогноз 5. Урок 10 — генератор

```python
doubled = (bill * 2 for bill in [100, 200])
print(list(doubled), list(doubled))
```

<details>
<summary>Відповідь</summary>

`[200, 400] []` — генератор вичерпується після першого проходу.

</details>

### Прогноз 6. Урок 13 — try / except / else / finally

```python
try:
    int("7.5")
except ValueError:
    print("A")
else:
    print("B")
finally:
    print("C")
```

<details>
<summary>Відповідь</summary>

`A`, потім `C`: `else` лише без винятку, `finally` — завжди.

</details>

### Прогноз 7. Урок 14 — JSON

```python
import json

print(json.loads(json.dumps({7: ("пт", 540)})))
```

<details>
<summary>Відповідь</summary>

`{'7': ['пт', 540]}` — ключ став рядком, кортеж — списком.

</details>

```python
%%writefile my-capstone/rules.py
"""Частина 1. Правила кафе (урок 12)."""
from models import DAYS


def meal_type(hour):
    """Прийом їжі за годиною: 11–15 → "обід", 17–23 → "вечеря", решта → "інше"."""
    # YOUR CODE HERE
    # BEGIN SOLUTION
    if 11 <= hour <= 15:
        return "обід"
    if 17 <= hour <= 23:
        return "вечеря"
    return "інше"
    # END SOLUTION


def day_name(time):
    """Короткий день тижня з DAYS: datetime(2024, 7, 19, 18, 30) → "пт"."""
    # YOUR CODE HERE
    # BEGIN SOLUTION
    return DAYS[time.weekday()]
    # END SOLUTION
```
