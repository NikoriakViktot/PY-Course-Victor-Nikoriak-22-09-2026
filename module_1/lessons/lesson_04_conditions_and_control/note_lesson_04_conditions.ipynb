## 📖 CONCEPT

### 1. `bool` — підклас `int`

| Значення | Числовий еквівалент |
|---|---|
| `True` | `1` |
| `False` | `0` |

Тому арифметика з `bool` цілком законна: `True + True` → `2`.

### 2. Truthy / Falsy

Кожен об'єкт неявно оцінюється як `True` або `False` в умові.

**Завжди Falsy:** `0`, `0.0`, `""`, `[]`, `{}`, `None`. **Усе інше** — Truthy.

### 3. Логічні оператори та пріоритет

`not` → `and` → `or` (зліва направо всередині одного рівня). Тому `not x and y or z` читається як `((not x) and y) or z`.

**Short-circuit:** Python зупиняється, щойно результат стає очевидним — `True or f()` НЕ викликає `f()`.

### 4. `None`

Особливий об'єкт (один-єдиний у всій програмі): означає «значення відсутнє». Функція без `return` завжди повертає `None`. Перевіряти через `is None`, не `== None`.

### 5. `if` / `elif` / `else`

Python перевіряє умови **зверху вниз** і виконує **перший** блок, де умова `True`. Тому умови розміщують від **конкретніших** до **загальніших** — інакше загальніша умова "з'їсть" усі випадки раніше, ніж дійде черга до точнішої.

### 6. `while`

Підходить, коли кількість повторень наперед невідома. **Завжди** зміни умову в тілі циклу — інакше нескінченний цикл.

### 7. `break` / `continue`

`break` — вийти з циклу повністю. `continue` — пропустити лише поточну ітерацію, цикл продовжується.

> Тут навмисно немає `if __name__ == "__main__"` — цю конструкцію ми вже трохи бачили на Уроці 2 і повністю розберемо на Уроці 12 («Модулі»), коли буде зрозуміло, навіщо взагалі ховати код від імпорту.

````python
# bool — підклас int
print("True + True  =", True + True)
print("isinstance(True, int) =", isinstance(True, int))

print()
for v in [0, 0.0, "", [], {}, None, 1, -1, "0", [0], " "]:
    print(f"  bool({v!r:8}) = {bool(v)}")
````

Що саме буде надруковано? Запиши точну фразу.

<details>

### Крок 3 — зміна вимог

Додай **четверту** категорію: `humidity` рівно `100` (або більше) — `"насичено"` (замість `"волого"`). Онов клітинку крок 2 логіку сам(-а), або допиши тут заново.

````python
def humidity_level(humidity):
    # BEGIN SOLUTION
    if humidity < 30:
        return "сухо"
    elif humidity <= 60:
        return "нормально"
    elif humidity < 100:
        return "волого"
    else:
        return "насичено"
    # END SOLUTION

for h in [10, 45, 80, 100, 120]:
    print(h, "->", humidity_level(h))

assert humidity_level(100) == "насичено"
assert humidity_level(99) == "волого"
print("OK")
````

### Крок 4 — самостійна задача (Task 4 у знайомому форматі)

Напиши власну класифікацію **індексу UV** (0–11+): `0-2` → `"низький"`, `3-5` → `"помірний"`, `6-7` → `"високий"`, `8-10` → `"дуже високий"`, `11+` → `"екстремальний"`. Порядок умов має значення — подумай, з якого боку почати.

````python
def uv_level(uv):
    # BEGIN SOLUTION
    if uv >= 11:
        return "екстремальний"
    elif uv >= 8:
        return "дуже високий"
    elif uv >= 6:
        return "високий"
    elif uv >= 3:
        return "помірний"
    else:
        return "низький"
    # END SOLUTION

for uv in [0, 2, 4, 6, 9, 11, 15]:
    print(uv, "->", uv_level(uv))

assert uv_level(0) == "низький"
assert uv_level(5) == "помірний"
assert uv_level(7) == "високий"
assert uv_level(10) == "дуже високий"
assert uv_level(11) == "екстремальний"
print("OK")
````

## 🔄 TRANSFER — та сама структура, інші дані

У прикладах вище класифікація завжди йшла за **діапазонами значення** (погода, вологість, UV). Тепер — задача з **іншою поверхневою деталлю**, але тією самою структурою рішення: вартість доставки залежно від ваги посилки (кг):

- до 1 кг включно → `50` грн
- більше 1 і до 5 включно → `120` грн
- більше 5 і до 20 включно → `250` грн
- більше 20 → `"уточнюйте вартість окремо"` (рядок, не число!)

Зверни увагу: результат не завжди число — це навмисно, щоб не можна було просто скопіювати структуру попередніх прикладів не думаючи.

````python
def shipping_cost(weight_kg):
    # BEGIN SOLUTION
    if weight_kg <= 1:
        return 50
    elif weight_kg <= 5:
        return 120
    elif weight_kg <= 20:
        return 250
    else:
        return "уточнюйте вартість окремо"
    # END SOLUTION

for w in [0.5, 1, 3, 5, 10, 20, 25]:
    print(w, "кг ->", shipping_cost(w))

assert shipping_cost(1) == 50
assert shipping_cost(5) == 120
assert shipping_cost(20) == 250
assert shipping_cost(21) == "уточнюйте вартість окремо"
print("OK")
````

### Самостійна задача на `while`

Симулюй введення PIN-коду: PIN має бути рівно 4 цифри. Використай список `attempts` замість справжнього `input()` (як на Уроці 2 не було `input()` для файлів — тут та сама техніка симуляції).

````python
attempts = ["12", "abcd", "1234"]
attempt_index = 0

# BEGIN SOLUTION
while True:
    user_input = attempts[attempt_index]
    attempt_index += 1
    print(f"Введено: {user_input!r}", end=" -> ")
    if user_input.isdigit() and len(user_input) == 4:
        print("Прийнято:", user_input)
        break
    else:
        print("Не 4-значний PIN, спробуйте ще")
# END SOLUTION
````

## 🔁 `while … else` — упізнай, коли побачиш

Гілка `else` після `while` виконується, коли цикл завершився **сам** (умова стала хибною), і **не** виконується після `break`. Порівняй два запуски:

````python
for secret in [2, 9]:          # for лише перебирає два приклади (урок 6)
    print("Шукаємо", secret)
    n = 1
    while n <= 3:
        if n == secret:
            print("  знайшли на кроці", n)
            break
        n += 1
    else:
        print("  не знайшли за 3 кроки")
````

## 🧭 Аргументи командного рядка: `sys.argv`

`input()` питає дані **під час** роботи. `sys.argv` отримує їх **під час запуску**: `python grade.py 87`. Кожен елемент `sys.argv` — **рядок**, `sys.argv[0]` — ім'я файлу.

У ноутбуці немає терміналу, тому:

1. клітинка з `%%writefile назва.py` **записує** програму у файл (працює і в Jupyter, і в Colab);
2. функція `run(...)` нижче **запускає** цей файл тим самим Python, що й ноутбук, — так само, як `python назва.py аргументи` в терміналі, — і повертає, що програма надрукувала.

````python
import subprocess
import sys


def run(*args, stdin=""):
    """Запустити python-файл з аргументами, як у терміналі; повернути надрукований текст."""
    result = subprocess.run([sys.executable, *args], input=stdin,
                            capture_output=True, text=True)
    return result.stdout + result.stderr
````

### Розібраний приклад: `grade.py`

Порядок перевірок: спершу **кількість** аргументів (інакше `sys.argv[1]` не існує), потім **формат** (`.isdigit()`), і лише потім `int()` та логіка оцінки (A ≥ 90, B ≥ 75, C ≥ 60, інакше F; більше 100 — помилка).

````python
%%writefile grade.py
import sys

if len(sys.argv) != 2:
    print("Використання: python grade.py <бал>")
elif not sys.argv[1].isdigit():
    print("Бал має бути цілим числом, отримано:", sys.argv[1])
else:
    score = int(sys.argv[1])
    if score > 100:
        print("Бал не може бути більшим за 100")
    elif score >= 90:
        print("A")
    elif score >= 75:
        print("B")
    elif score >= 60:
        print("C")
    else:
        print("F")
````

### Вправа A. Зворотний відлік — `countdown.py`

`python countdown.py 3` друкує `3`, `2`, `1`, `Поїхали!` (кожне з нового рядка). Без аргументу, з двома аргументами чи не з числом — `Використання: python countdown.py <ціле число>`. Заміни `# YOUR CODE HERE` і запусти обидві клітинки.

````python
%%writefile countdown.py
import sys

# YOUR CODE HERE
# BEGIN SOLUTION
if len(sys.argv) != 2 or not sys.argv[1].isdigit():
    print("Використання: python countdown.py <ціле число>")
else:
    n = int(sys.argv[1])
    while n > 0:
        print(n)
        n -= 1
    print("Поїхали!")
# END SOLUTION
````

````python
print(run("countdown.py", "3"))
assert run("countdown.py", "3") == "3\n2\n1\nПоїхали!\n"
assert run("countdown.py", "0") == "Поїхали!\n"
usage = "Використання: python countdown.py <ціле число>\n"
assert run("countdown.py") == usage
assert run("countdown.py", "3", "4") == usage
assert run("countdown.py", "три") == usage
print("✅ Вправа A пройдена")
````

### Вправа B. Сума цифр — `digits.py`

`python digits.py 2026` → `Сума цифр: 10`. Остання цифра — `number % 10`, відкинути її — `number //= 10`, повторювати `while number > 0`. Перевірка аргументу — як у вправі A (`Використання: python digits.py <ціле число>`).

````python
%%writefile digits.py
import sys

# YOUR CODE HERE
# BEGIN SOLUTION
if len(sys.argv) != 2 or not sys.argv[1].isdigit():
    print("Використання: python digits.py <ціле число>")
else:
    number = int(sys.argv[1])
    total = 0
    while number > 0:
        total += number % 10
        number //= 10
    print("Сума цифр:", total)
# END SOLUTION
````

````python
print(run("digits.py", "2026"))
assert run("digits.py", "2026") == "Сума цифр: 10\n"
assert run("digits.py", "0") == "Сума цифр: 0\n"
assert run("digits.py", "99999") == "Сума цифр: 45\n"
assert run("digits.py", "-15") == "Використання: python digits.py <ціле число>\n"
print("✅ Вправа B пройдена")
````

### Вправа C. Вгадай число за N спроб — `guess.py`

- аргумент — кількість спроб (перевірка, як у вправі A: `Використання: python guess.py <кількість спроб>`);
- загадане число — `SECRET = 13` у коді;
- у `while` читай `input("Твоє число: ")`, очищай `.strip()`;
- не число → `Потрібне ціле число` і `continue` **без** витрати спроби;
- вгадав → `Вгадав!` і `break`; інакше підказка `Більше` / `Менше`;
- спроби скінчились → гілка `else` циклу: `Спроби закінчились. Було загадано: 13`.

`run(..., stdin="10\n15\n")` подає програмі введення — ніби користувач надрукував ці рядки. Сама підказка `Твоє число: ` виводиться без переносу, тому у виводі відповіді стоять на тому самому рядку.

````python
%%writefile guess.py
import sys

SECRET = 13

# YOUR CODE HERE
# BEGIN SOLUTION
if len(sys.argv) != 2 or not sys.argv[1].isdigit():
    print("Використання: python guess.py <кількість спроб>")
else:
    attempts = int(sys.argv[1])
    while attempts > 0:
        text = input("Твоє число: ").strip()
        if not text.isdigit():
            print("Потрібне ціле число")
            continue
        attempts -= 1
        guess = int(text)
        if guess == SECRET:
            print("Вгадав!")
            break
        elif guess < SECRET:
            print("Більше")
        else:
            print("Менше")
    else:
        print("Спроби закінчились. Було загадано:", SECRET)
# END SOLUTION
````

````python
print(run("guess.py", "3", stdin="10\nп'ять\n15\n13\n"))
print(run("guess.py", "2", stdin="1\n2\n"))
assert run("guess.py", "3", stdin="10\nп'ять\n15\n13\n").endswith("Вгадав!\n")
assert run("guess.py", "2", stdin="1\n2\n").endswith("Спроби закінчились. Було загадано: 13\n")
assert run("guess.py", "1", stdin="13\n").endswith("Вгадав!\n")
assert "Спроби закінчились" not in run("guess.py", "1", stdin="13\n")
assert run("guess.py") == "Використання: python guess.py <кількість спроб>\n"
print("✅ Вправа C пройдена")
````

## 🧯 Читаємо помилки
