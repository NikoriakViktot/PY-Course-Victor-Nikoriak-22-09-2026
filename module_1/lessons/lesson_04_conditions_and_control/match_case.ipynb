## 1. Від `elif` до `match`

Кафе «Смачно» працює в будні з 08:00 до 22:00, а у вихідні — з 10:00 до 23:00. Спершу — знайомим ланцюжком `if` / `elif` / `else`:

````python
day = "субота"

if day == "субота" or day == "неділя":
    print("Вихідний: 10:00–23:00")
elif day in ("понеділок", "вівторок", "середа", "четвер", "п'ятниця"):
    print("Будній: 08:00–22:00")
else:
    print("Невідомий день:", day)
````

Те саме через `match`:

- `match day:` — **яке** значення перевіряємо (пишемо один раз);
- `case "субота" | "неділя":` — гілка для цих значень, `|` означає «або»;
- `case _:` — «усе інше», роль `else`.

````python
day = "субота"

match day:
    case "субота" | "неділя":
        print("Вихідний: 10:00–23:00")
    case "понеділок" | "вівторок" | "середа" | "четвер" | "п'ятниця":
        print("Будній: 08:00–22:00")
    case _:
        print("Невідомий день:", day)
````

**🔮 Прогноз:** що виведе той самий `match`, якщо `day = "Неділя"` (з великої літери)?

<details>
<summary>Відповідь</summary>

`Невідомий день: Неділя`. `case "неділя"` порівнює рядки **точно**, як `==`: `"Неділя" == "неділя"` — `False`. Тому дані спершу очищають: `day.strip().lower()` — як з `input()` в уроці 4.

</details>

````python
day = "Неділя"

match day:
    case "субота" | "неділя":
        print("Вихідний: 10:00–23:00")
    case "понеділок" | "вівторок" | "середа" | "четвер" | "п'ятниця":
        print("Будній: 08:00–22:00")
    case _:
        print("Невідомий день:", day)
````

## 2. Як виконується `match`

1. Python обчислює значення після `match` **один раз**.
2. Перевіряє гілки `case` **зверху вниз**.
3. Виконує **першу**, що підійшла, і виходить з `match`. Решта гілок не перевіряється — `break`, як у `switch` інших мов, не потрібен.
4. Якщо не підійшла жодна і немає `case _`, **нічого не відбувається** — ні помилки, ні повідомлення.

Служба доставки повертає числовий код відповіді:

````python
status = 404

match status:
    case 200:
        print("Замовлення прийнято")
    case 404:
        print("Ресторан не знайдено")
    case 500:
        print("Помилка сервера, спробуй пізніше")
    case _:
        print("Невідомий код:", status)
````

**🔮 Прогноз:** що виведе цей код при `status = 302`?

```python
status = 302

match status:
    case 200:
        print("Замовлення прийнято")
    case 404:
        print("Ресторан не знайдено")

print("Кінець")
```

<details>
<summary>Відповідь</summary>

Лише `Кінець`. Жодна гілка не підійшла, а `case _` немає — `match` мовчки пропускається. Тому майже завжди додавай `case _`, щоб неочікуване значення не загубилось.

</details>

````python
status = 302

match status:
    case 200:
        print("Замовлення прийнято")
    case 404:
        print("Ресторан не знайдено")

print("Кінець")
````

## 3. Пастки: голе ім'я в `case`

У `case` **літерал** (`"субота"`, `404`) порівнюється зі значенням. А **голе ім'я** (`case other:`, `case FREE:`) — ні: воно **захоплює** будь-яке значення і записує його в цю змінну. Така гілка спрацьовує завжди.

Якщо після неї є інші гілки, Python навіть не запустить код:

````python
code_text = """
match status:
    case other:
        print(other)
    case 200:
        print("OK")
"""

try:
    compile(code_text, "<cell>", "exec")   # лише перевіряємо синтаксис, не виконуємо
except SyntaxError as e:
    print("SyntaxError:", e.msg)
````

Підступніше, коли така гілка **остання** — тоді помилки немає. Хочемо перевірити, чи доставка безкоштовна, і пишемо «константу» в `case`:

````python
FREE = "безкоштовно"
delivery = "60 грн"

match delivery:
    case FREE:
        print("Спрацювало:", FREE)

print("FREE тепер:", FREE)
````

Гілка спрацювала для `"60 грн"`, а змінна `FREE` **перезаписалась**. `case FREE:` не порівнює з `FREE`, а захоплює значення в змінну `FREE`.

**Правило:** у `case` пиши літерали (`case "безкоштовно":`), а для «усього іншого» — `case _:`. Порівняння зі змінною роби через `if` або guard (розділ 4).

---
## 4. Умова в гілці: guard

Після зразка можна додати `if` — **guard** (охоронець). Гілка спрацює, лише якщо підійшов зразок **і** умова істинна. Кафе садить гостей за кількістю:

````python
for guests in [0, 2, 5, 12]:
    match guests:
        case 0:
            print(guests, "→ стіл вільний")
        case 1 | 2:
            print(guests, "→ столик на двох")
        case n if n <= 6:
            print(guests, f"→ стіл на {n}")
        case _:
            print(guests, "→ банкетна зала")
````

Цикл `for` лише перебирає кілька прикладів (детально — урок 6).

`case n if n <= 6:` — ім'я `n` захоплює число, а guard пропускає лише `n <= 6`. Для `12` guard хибний, тому Python іде далі — до `case _`.

Якщо **усі** гілки — це діапазони (`>= 500`, `>= 200`…), звичайний `if` / `elif` читається простіше. `match` сильний там, де більшість гілок — конкретні значення.

## 5. Забігаємо наперед: зразки за формою

Головна сила `match` — він перевіряє не лише значення, а **форму даних** і одразу витягає з них частини. Повністю це знадобиться після уроків про колекції, але подивитись варто вже зараз.

### 5.1. Команди офіціанта: список слів (урок 5)

`text.split()` розбиває рядок на слова (методи рядків — урок 3). `case ["додай", item]` спрацює, коли слів **рівно два** і перше — `"додай"`; друге слово потрапить у `item`:

````python
for text in ["додай кава", "рахунок", "знижка STUDENT10", "додай", "готово зараз"]:
    match text.split():
        case ["додай", item]:
            print(f"{text!r:22} → додаємо {item}")
        case ["рахунок"]:
            print(f"{text!r:22} → показуємо рахунок")
        case ["знижка", code]:
            print(f"{text!r:22} → код знижки {code}")
        case ["готово"]:
            print(f"{text!r:22} → закриваємо замовлення")
        case _:
            print(f"{text!r:22} → не розумію команду")
````

`"додай"` без назви — одне слово, тож `["додай", item]` не підходить. `"готово зараз"` — два слова, а `["готово"]` чекає рівно одне. Через `if` довелося б окремо перевіряти і кількість слів, і перше слово.

### 5.2. Замовлення-словник (урок 6)

`case {"type": "delivery", "address": address}` спрацює, якщо в словнику є ключ `"type"` зі значенням `"delivery"` і ключ `"address"`; його значення потрапить у змінну `address`. Зайві ключі (`"total"`) не заважають.

````python
orders = [
    {"type": "delivery", "address": "Хрещатик, 1", "total": 540},
    {"type": "hall", "table": 7},
    {"type": "pickup"},
]

for order in orders:
    match order:
        case {"type": "delivery", "address": address}:
            print("Доставка на адресу:", address)
        case {"type": "hall", "table": table}:
            print("Столик №", table)
        case _:
            print("Невідомий тип:", order)
````

### 5.3. Об'єкти класів (урок 19)

Зразок `Point(x=0, y=y)` перевіряє, що об'єкт — `Point` і його `x` дорівнює `0`, а `y` захоплює. `@dataclass` створює простий клас з полями; класи розбиратимемо в модулі 2.

````python
from dataclasses import dataclass


@dataclass
class Point:
    x: int
    y: int


for point in [Point(0, 0), Point(0, 5), Point(3, 7)]:
    match point:
        case Point(x=0, y=0):
            print(point, "→ кафе (початок координат)")
        case Point(x=0, y=y):
            print(point, f"→ на головній вулиці, {y} км від кафе")
        case Point():
            print(point, "→ десь у місті")
````

## 6. `match` чи `if` / `elif`

| | `if` / `elif` / `else` | `match` / `case` |
|---|---|---|
| Версія Python | будь-яка | 3.10+ |
| Що перевіряє | будь-яку умову: `>=`, `and`, виклики функцій | збіг із зразком: значення, форма списку/словника, клас |
| «Усе інше» | `else` | `case _` |
| Виконується | перша істинна гілка | перша гілка, що підійшла |
| Витягти частини даних | окремими рядками після перевірки | прямо в зразку: `["додай", item]` |
| Пастка | порядок діапазонів (урок 4) | голе ім'я захоплює все |

**Коротко:** діапазони й складні умови — `if`; вибір з конкретних значень або розбір даних за формою — `match`.

# 7. 🛠 Вправи

У кожній вправі заміни `# YOUR CODE HERE` на свій код і запусти клітинку. Якщо все правильно — побачиш ✅. Використовуй `match`.

### Вправа 1. Години роботи

Функція отримує день тижня так, як його ввів користувач (з пробілами, будь-яким регістром) і повертає рядок з годинами:

- будні — `"08:00–22:00"`;
- субота й неділя — `"10:00–23:00"`;
- інше — `"Невідомий день"`.

````python
def opening_hours(day):
    # YOUR CODE HERE
    # BEGIN SOLUTION
    match day.strip().lower():
        case "субота" | "неділя":
            return "10:00–23:00"
        case "понеділок" | "вівторок" | "середа" | "четвер" | "п'ятниця":
            return "08:00–22:00"
        case _:
            return "Невідомий день"
    # END SOLUTION


assert opening_hours("середа") == "08:00–22:00"
assert opening_hours("  Неділя ") == "10:00–23:00"
assert opening_hours("СУБОТА") == "10:00–23:00"
assert opening_hours("свято") == "Невідомий день"
print("✅ Вправа 1 пройдена")
````

### Вправа 2. Спосіб оплати

Каса приймає українські й англійські назви. Поверни:

- `"картка"` або `"card"` → `"Оплата карткою"`;
- `"готівка"` або `"cash"` → `"Оплата готівкою"`;
- інше → `"Спосіб оплати не підтримується: <що ввели>"`.

````python
def payment_message(method):
    # YOUR CODE HERE
    # BEGIN SOLUTION
    match method:
        case "картка" | "card":
            return "Оплата карткою"
        case "готівка" | "cash":
            return "Оплата готівкою"
        case _:
            return f"Спосіб оплати не підтримується: {method}"
    # END SOLUTION


assert payment_message("card") == "Оплата карткою"
assert payment_message("готівка") == "Оплата готівкою"
assert payment_message("біткоїн") == "Спосіб оплати не підтримується: біткоїн"
print("✅ Вправа 2 пройдена")
````

### Вправа 3. Розмір порції (guard)

За вагою порції в грамах поверни розмір:

- `0` → `"Порожня тарілка"`;
- від 1 до 250 → `"Мала"`;
- від 251 до 400 → `"Середня"`;
- більше 400 → `"Велика"`;
- від'ємна вага → `"Помилка ваги"`.

Використай `case 0:` для нуля і guards для решти.

````python
def portion_size(grams):
    # YOUR CODE HERE
    # BEGIN SOLUTION
    match grams:
        case 0:
            return "Порожня тарілка"
        case g if g < 0:
            return "Помилка ваги"
        case g if g <= 250:
            return "Мала"
        case g if g <= 400:
            return "Середня"
        case _:
            return "Велика"
    # END SOLUTION


assert portion_size(0) == "Порожня тарілка"
assert portion_size(250) == "Мала"
assert portion_size(251) == "Середня"
assert portion_size(400) == "Середня"
assert portion_size(401) == "Велика"
assert portion_size(-5) == "Помилка ваги"
print("✅ Вправа 3 пройдена")
````

### Вправа 4. Команди офіціанта (забігаємо в урок 5)

Розбий команду на слова через `.split()` і розбери `match`:

- `"додай <позиція>"` → `"Додано: <позиція>"`;
- `"прибери <позиція>"` → `"Прибрано: <позиція>"`;
- `"рахунок"` → `"Рахунок"`;
- усе інше (зокрема порожній рядок) → `"Невідома команда"`.

````python
def waiter_command(text):
    # YOUR CODE HERE
    # BEGIN SOLUTION
    match text.split():
        case ["додай", item]:
            return f"Додано: {item}"
        case ["прибери", item]:
            return f"Прибрано: {item}"
        case ["рахунок"]:
            return "Рахунок"
        case _:
            return "Невідома команда"
    # END SOLUTION


assert waiter_command("додай круасан") == "Додано: круасан"
assert waiter_command("прибери кава") == "Прибрано: кава"
assert waiter_command("  рахунок ") == "Рахунок"
assert waiter_command("додай") == "Невідома команда"
assert waiter_command("") == "Невідома команда"
assert waiter_command("рахунок будь ласка") == "Невідома команда"
print("✅ Вправа 4 пройдена")
````
