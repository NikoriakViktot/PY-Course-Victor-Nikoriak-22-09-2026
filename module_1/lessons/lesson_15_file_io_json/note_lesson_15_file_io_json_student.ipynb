<details>
<summary>Відповідь</summary>

`'2024-07-19 18:30;540.00;50;2\n'` — у кожного рядка в кінці символ переходу `\n`. Його прибирають `strip()`.

</details>

````python
def read_kasa(path):
    """Непорожні рядки файлу каси без символу нового рядка."""
    lines = []
    with open(path, encoding="utf-8") as file:
        for line in file:
            line = line.strip()
            if line:
                lines.append(line)
    return lines


lines = read_kasa("kasa_2024_07.txt")
print(len(lines), lines[-1])
````

## 2. Розбір рядків — код з уроку 14 без змін

````python
from datetime import datetime
from typing import NamedTuple


class RawOrder(NamedTuple):
    total_bill: float
    tip: float
    size: int
    timestamp: datetime


def parse_line(line):
    """Рядок каси -> RawOrder. Зіпсований рядок -> ValueError з поясненням."""
    fields = line.split(";")
    if len(fields) != 4:
        raise ValueError(f"очікували 4 поля, а маємо {len(fields)}")
    time_text, bill_text, tip_text, size_text = fields
    timestamp = datetime.strptime(time_text, "%Y-%m-%d %H:%M")
    bill, tip, size = float(bill_text), float(tip_text), int(size_text)
    if bill <= 0:
        raise ValueError(f"сума чека має бути більшою за 0, а маємо {bill}")
    if tip < 0:
        raise ValueError(f"чайові не можуть бути від'ємними: {tip}")
    if size < 1:
        raise ValueError(f"гостей має бути хоча б один, а маємо {size}")
    return RawOrder(bill, tip, size, timestamp)


def load_orders(lines):
    """Правильні чеки і список пояснень до пропущених рядків."""
    orders = []
    errors = []
    for number, line in enumerate(lines, start=1):
        try:
            orders.append(parse_line(line))
        except ValueError as error:
            errors.append(f"рядок {number}: {error}")
    return orders, errors
````

## 8. Розібраний приклад: місячний звіт

`build_report` лише рахує; `month_report` читає й пише файли; `main` розмовляє з людиною.

````python
DAYS = ("пн", "вт", "ср", "чт", "пт", "сб", "нд")


def meal_type_from_hour(hour):
    if 11 <= hour <= 15:
        return "обід"
    if 17 <= hour <= 23:
        return "вечеря"
    return "інше"
````

````python
def build_report(config, period, orders, errors):
    """Звіт за місяць — лише типи, які розуміє JSON."""
    revenue = sum(order.total_bill for order in orders)
    average = revenue / len(orders) if orders else 0.0
    by_time = Counter(meal_type_from_hour(order.timestamp.hour) for order in orders)
    first = min(order.timestamp for order in orders).isoformat() if orders else None
    return {
        "cafe": config["cafe"],
        "currency": config["currency"],
        "period": period,
        "orders": len(orders),
        "skipped": len(errors),
        "revenue": round(revenue, 2),
        "average": round(average, 2),
        "by_time": dict(by_time.most_common()),
        "first_order": first,
    }


def month_report(config, year, month):
    names = {key: config[key].format(year=year, month=month)
             for key in ("kasa_file", "report_file", "errors_file")}
    orders, errors = load_orders(read_kasa(names["kasa_file"]))
    report = build_report(config, f"{year}-{month:02d}", orders, errors)

    with open(names["report_file"], "w", encoding="utf-8") as file:
        json.dump(report, file, ensure_ascii=False, indent=2)
    with open(names["errors_file"], "w", encoding="utf-8") as file:
        for message in errors:
            print(message, file=file)
    log_run(f"{report['period']}: прийнято {report['orders']}, пропущено {report['skipped']}",
            config["log_file"])
    return names["report_file"]


def main(year, month):
    with open("cafe_config.json", encoding="utf-8") as file:
        config = json.load(file)
    try:
        report_file = month_report(config, year, month)
    except FileNotFoundError as error:
        print("Немає файлу каси:", error.filename)
        return 1
    print("Звіт збережено:", report_file)
    return 0


main(2024, 7)
main(2024, 8)
````

## ➕ Поглиблення: усі режими `open()`

Для звіту вистачило `r`, `w`, `a`. Цей розділ — для охочих: як не затерти готовий файл, як читати й писати одним відкриттям і як працювати з байтами.

**4 базові режими** (базова літера в режимі — рівно одна):

| Режим | Файл є | Файлу немає | Курсор |
|---|---|---|---|
| `r` | читає з початку | `FileNotFoundError` | на початку |
| `w` | **стирає** одразу | створює | на початку |
| `a` | дописує в кінець | створює | у кінці |
| `x` | `FileExistsError` | створює | на початку |

**3 модифікатори:** `t` — текст, `str` (за замовчуванням); `b` — байти, `bytes`, без `encoding`; `+` — і читати, і писати.

**16 комбінацій** — 8 текстових і 8 двійкових:

| Текстові | Двійкові | Читати | Писати | Стирає файл |
|---|---|:---:|:---:|:---:|
| `r` / `rt` | `rb` | ✅ | — | — |
| `w` / `wt` | `wb` | — | ✅ | ✅ |
| `a` / `at` | `ab` | — | ✅ в кінець | — |
| `x` / `xt` | `xb` | — | ✅ | файлу не було |
| `r+` / `r+t` | `r+b` | ✅ | ✅ поверх | — |
| `w+` / `w+t` | `w+b` | ✅ | ✅ | ✅ |
| `a+` / `a+t` | `a+b` | ✅ | ✅ в кінець | — |
| `x+` / `x+t` | `x+b` | ✅ | ✅ | файлу не було |

Порядок літер не важливий: `"rb+"` = `"r+b"`.

**Прогноз:** у яких із восьми текстових режимів файл зі словом `"дані"` стане порожнім **одразу після відкриття**, ще до запису? Перевір клітинкою нижче.

````python
for mode in ["r", "w", "a", "x", "r+", "w+", "a+", "x+"]:
    if "x" in mode:
        Path("probe.txt").unlink(missing_ok=True)   # x потребує, щоб файлу не було
    else:
        Path("probe.txt").write_text("дані", encoding="utf-8")
    with open("probe.txt", mode, encoding="utf-8") as file:
        cursor = "початок" if file.tell() == 0 else "кінець"
        content = Path("probe.txt").read_text(encoding="utf-8")
        print(f"{mode:<3} читати: {file.readable()!s:<5} писати: {file.writable()!s:<5} "
              f"курсор: {cursor:<7} у файлі: {content!r}")
````

<details>
<summary>Відповідь</summary>

`w` і `w+` стирають наявний файл одразу. `x` і `x+` теж дають порожній файл, але лише тому, що створюють новий. `+` змінює тільки «читати / писати»; що станеться з файлом, вирішує базова літера.

</details>

### `x`: не затерти готовий звіт

Звіт за липень уже відправили. Другий запуск з `"w"` тихо перезаписав би його. З `"x"` — `FileExistsError`.

````python
Path("report_x.json").unlink(missing_ok=True)   # починаємо з чистого стану
report = {"period": "2024-07", "orders": 4, "revenue": 3040.0}

for attempt in (1, 2):
    try:
        with open("report_x.json", "x", encoding="utf-8") as file:
            json.dump(report, file, ensure_ascii=False)
        print(f"Спроба {attempt}: звіт створено")
    except FileExistsError as error:
        print(f"Спроба {attempt}: звіт уже є —", error.filename)
````

### `r+`: пише поверх, а не вставляє

**Прогноз:** у файлі `зміна: відкрита`. Перемотуємо курсор на початок і пишемо `зміна: закрита`. Що буде у файлі?

````python
Path("shift.txt").write_text("зміна: відкрита", encoding="utf-8")

with open("shift.txt", "r+", encoding="utf-8") as file:
    print("прочитано:", file.read())
    file.seek(0)                     # курсор назад на початок
    file.write("зміна: закрита")
    # file.truncate()                # розкоментуй — і хвоста не буде

print("у файлі:  ", Path("shift.txt").read_text(encoding="utf-8"))
````

<details>
<summary>Відповідь</summary>

`зміна: закритаа`. `r+` не очищає файл: новий текст ліг поверх старого, а старий був на літеру довший. Хвіст обрізає `file.truncate()`. Простіше: прочитати з `"r"`, змінити в пам'яті, записати з `"w"`.

</details>

### `w+` і `a+`: де курсор

Після `write` курсор стоїть за записаним текстом — `read()` нічого не знайде, поки не зробиш `seek(0)`. У `a+` курсор одразу в кінці, а запис іде в кінець навіть після `seek(0)`.

````python
with open("shift.txt", "w+", encoding="utf-8") as file:
    file.write("зміна: закрита")
    print("w+ після write:  ", repr(file.read()))
    file.seek(0)
    print("w+ після seek(0):", repr(file.read()))

Path("runs_plus.log").write_text("2024-06: прийнято 7\n", encoding="utf-8")
with open("runs_plus.log", "a+", encoding="utf-8") as file:
    print("a+ одразу:       ", repr(file.read()))
    file.seek(0)
    print("a+ після seek(0):", repr(file.read()))
    file.seek(0)
    file.write("2024-07: прийнято 4\n")   # однаково піде в кінець

print(Path("runs_plus.log").read_text(encoding="utf-8"), end="")
````

### `b`: байти замість тексту

Без `b` Python перетворює байти з диска на літери за `encoding`. З `b` — ні: отримуєш `bytes`. Потрібно для картинок, PDF, архівів, копії файлу «як є».

````python
Path("menu.txt").write_text("Кава 60", encoding="utf-8")

with open("menu.txt", encoding="utf-8") as file:
    text = file.read()
with open("menu.txt", "rb") as file:
    data = file.read()

print(type(text).__name__, len(text), repr(text))
print(type(data).__name__, len(data), data)   # кирилична літера в UTF-8 — 2 байти
print(data.decode("utf-8"))

try:
    open("menu.txt", "rb", encoding="utf-8")
except ValueError as error:
    print("ValueError:", error)

try:
    with open("menu_copy.bin", "wb") as file:
        file.write("Кава")                    # треба "Кава".encode("utf-8")
except TypeError as error:
    print("TypeError:", error)
````
