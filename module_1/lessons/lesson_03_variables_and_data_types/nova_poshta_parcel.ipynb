На склад Нової Пошти приїхала коробка без зворотної адреси. На ній — чотири порожні місця для пломб. Склад відкриє коробку лише тоді, коли **кожен із чотирьох відділів** перевірить свою частину даних і поставить пломбу.

```text
                 📦 КОРОБКА

 відділ 1   відділ 2    відділ 3     відділ 4
 Адресний   Ваговий    Логістика   Відправлення
```

Якщо хоча б один відділ помилиться — коробка лишиться закритою.

### Як грати

**На занятті.** Кожна група відкриває цей ноутбук і розв'язує **свій** відділ.

1. Запустіть клітинку «🔒 Система Нової Пошти» (один раз).
2. Прочитайте завдання свого відділу, замініть `None` на відповідь і запустіть клітинку.
3. Якщо відповідь правильна, система видасть **код пломби** — двоцифрове число. Лідер групи називає його викладачу.
4. Викладач вписує всі чотири коди в клітинку «🏬 Склад».

**Самостійно (удома).** Розв'яжи всі чотири відділи по черзі. Коди пломб збережуться у змінних `SEAL_1` … `SEAL_4`, і склад можна відкрити одним рядком.

> Системна клітинка не зберігає правильних відповідей відкритим текстом, тож підглянути не вийде. Але вона підкаже, якщо помилка типова.

````python
# @title 🔒 Система Нової Пошти — натисніть ▶ один раз
# Не змінюйте цю клітинку: тут перевірка відділів і замок складу.
import hashlib

_SALT = "nova-poshta-l3"
_ANSWERS = {
    1: "cdbee5544c4b1f834f1c6b8ff824fc624fbb05a262ffea8afa284ea43b87c11b",
    2: "f5db34d56ca6094a61b5e020e3c613e0dacbf4ba85cac8d469cbcb14af6651a6",
    3: "a46386a0ce43461cc8dd4ea0d50882f869cb939d0516fd93a9666ed35dfcbd15",
    4: "1d18b970c4f0f25f7f6ec86c5b54cef4be6ed91e856f378452ba8c3a3d595153",
}
_SEALS = {
    1: "391244f1a9ae13bd077bb7deeedd222f2450ba6adfbcf6901c47306f2440823a",
    2: "48ffabf2d637bf37b4dd304b07f8c4993d891385bec28cab162202d96906b8c9",
    3: "17dee0de5dfae4325957fa224d24fc752ccea6f63b50243928408038c14eeb01",
    4: "13d0c97cd18257b184e373e7daac2008b462e952b73c6d2980b451f6623f5d0d",
}
_DEPARTMENTS = {1: "Адресний відділ", 2: "Ваговий відділ", 3: "Логістика", 4: "Відділ відправлення"}
_CODES = (1050, 1080, 1111, 1074)


def _digest(text):
    return hashlib.sha256((_SALT + text).encode("utf-8")).hexdigest()


def _fingerprint(answer):
    return _digest(type(answer).__name__ + ":" + repr(answer))


def _seal(n, answer):
    return int(_digest("seal" + str(n) + _fingerprint(answer))[:8], 16) % 90 + 10


def _hint(n, answer):
    if n == 1:
        if answer == 200:
            return "tracking не стежить за box_id. Рядок box_id = 200 прив'язав до box_id нове число, а tracking досі вказує на старе."
        if isinstance(answer, str):
            return "Потрібне число, а не текст. Прибери лапки."
        return "Пройди код рядок за рядком: на яке значення вказує tracking одразу після tracking = box_id?"
    if n == 2:
        if answer == "25":
            return "Це все ще рядок \"25\". Система приймає лише int — перетвори його через int()."
        if isinstance(answer, float):
            return "Вага має бути цілим числом int, а не float."
        return "Перевір тип відповіді через type() і значення через print()."
    if n == 3:
        if answer == 60:
            return "Схоже, додавання спрацювало раніше за множення. У Python спершу множення, потім додавання."
        if isinstance(answer, str):
            return "Вартість — число. Прибери лапки."
        return "Підстав числа у формулу «база + вага × 2» і пам'ятай про порядок операцій."
    if n == 4:
        if isinstance(answer, int):
            return "Це число, а потрібен текст. Застосуй chr() до кожного коду і склей літери через +."
        if isinstance(answer, str) and len(answer) != 4:
            return "Кодів чотири, отже й літер має бути чотири — по одній на кожен код."
        return "Перевір кожен код окремо, наприклад print(chr(1050))."
    return ""


_MATERIALS = "6ee31ef06c8d8740f53e2c2811913900af93526c764b0b5bfb2d2dedaa0b8edbe27ae3779666c41c3e8c8012e4d6224c06f97491b67aa89672b2c2011923c01e02c36559310a9f920bc0c0e7c43739f0abd1d2d18ba887"


def _materials_link(seals):
    """Посилання на матеріали розшифровується лише правильними пломбами."""
    seed = "-".join(str(seals[n]) for n in (1, 2, 3, 4))
    data = bytes.fromhex(_MATERIALS)
    stream = b""
    i = 0
    while len(stream) < len(data):
        stream += hashlib.sha256((_SALT + "drive" + seed + str(i)).encode("utf-8")).digest()
        i += 1
    return bytes(a ^ b for a, b in zip(data, stream)).decode("utf-8")


def _show_materials(url):
    try:
        from IPython.display import HTML, display
    except ImportError:
        print("📚 Матеріали уроку:", url)
        return
    display(HTML(
        f'<a href="{url}" target="_blank" style="display:inline-block; padding:16px 28px; '
        'font-size:20px; color:white; background:#27ae60; border-radius:12px; '
        'text-decoration:none;">📚 Матеріали уроку</a>'
    ))


def check_department(n, answer):
    """Перевіряє відповідь відділу n. Якщо все правильно — повертає код пломби."""
    name = _DEPARTMENTS[n]
    if answer is None:
        print(f"✏️ {name}: впишіть відповідь замість None і запустіть клітинку ще раз.")
        return None
    if _fingerprint(answer) == _ANSWERS[n]:
        seal = _seal(n, answer)
        print(f"✅ {name}: дані перевірено.")
        print(f"🔏 Пломба відділу {n}: {seal}")
        print("   Лідер групи — назвіть цей код викладачу.")
        return seal
    print(f"❌ {name}: коробку не пропущено.")
    print("💡 " + _hint(n, answer))
    return None


def open_box(seal_1=None, seal_2=None, seal_3=None, seal_4=None):
    """Склад: відкриває коробку, коли всі чотири пломби правильні."""
    seals = {1: seal_1, 2: seal_2, 3: seal_3, 4: seal_4}
    print("🏬 Склад Нової Пошти перевіряє пломби...\n")
    wrong = []
    for n, seal in seals.items():
        if seal is not None and _digest("seal-check" + str(n) + str(seal)) == _SEALS[n]:
            print(f"   ✅ відділ {n} — {_DEPARTMENTS[n]}")
        else:
            print(f"   ❌ відділ {n} — {_DEPARTMENTS[n]}: пломби немає або вона неправильна")
            wrong.append(n)
    print()
    if wrong:
        print("🔒 Коробка залишається закритою.")
        print("   Перевірте відділи:", ", ".join(str(n) for n in wrong))
        return False
    city = chr(_CODES[0]) + chr(_CODES[1]) + chr(_CODES[2]) + chr(_CODES[3])
    print("🔓 КОРОБКУ ВІДКРИТО!")
    print("📍 Місто відправлення:", city)
    print()
    print("Усередині — лист:")
    print("   «Дякуємо всім відділам! Ви щойно:")
    print("    • простежили, куди вказують імена після переприсвоювання;")
    print("    • перетворили текст \"25\" на число 25;")
    print("    • порахували вартість з правильним порядком операцій;")
    print("    • прочитали назву міста з кодів символів через chr().")
    print("    Це і є змінні та базові типи даних.»")
    print()
    print("🎁 А ще в коробці — ключ до матеріалів уроку:")
    _show_materials(_materials_link(seals))
    return True


print("🔧 Система Нової Пошти готова ✅")
````
